# Bagian 1: Penjelasan Konsep

## 1.1 Definisi dan Posisi Preprocessing

* Preprocessing, dalam konteks scikit-learn, didefinisikan sebagai transformasi numerik terstruktur yang mengubah data mentah (hasil cleaning) menjadi bentuk yang dapat dikonsumsi oleh algoritma ML.
* Preprocessing bukan sinonim dari data cleaning:

| Aspek           | Data Cleaning (Fase 1, Pandas)                                              | Preprocessing (Fase 3, Scikit-learn)                                              |
| --------------- | --------------------------------------------------------------------------- | --------------------------------------------------------------------------------- |
| Tujuan          | Memperbaiki kualitas data (duplikat, error, format tidak konsisten)         | Mengubah data bersih menjadi representasi yang sesuai untuk algoritma             |
| Sifat keputusan | Berbasis domain/bisnis, misalnya menentukan apakah outlier valid atau error | Berbasis kebutuhan matematis algoritma, misalnya skala fitur dan tipe input       |
| Contoh operasi  | `drop_duplicates()`, standardisasi format tanggal                           | `StandardScaler`, `OneHotEncoder`, `SimpleImputer`                                |
| Kapan dilakukan | Sebelum split, untuk memahami dan memperbaiki kualitas dataset              | Setelah split, dengan preprocessing dipelajari pada train lalu diterapkan ke test |


* Preprocessing berada di antara dua tahap: data yang sudah bersih (output Fase 1) dan model yang siap dilatih (input .fit() pada predictor).
* Alasan preprocessing diperlukan secara fundamental: algoritma ML pada dasarnya adalah operasi matematis atas angka — algoritma tidak dapat memproses teks kategorikal secara langsung, dan sebagian besar algoritma sensitif terhadap skala numerik yang tidak seragam antar fitur.
