# Bagian 1 — Penjelasan

## 1.1 Apa itu EDA dan Tujuannya

EDA (Exploratory Data Analysis): Proses sistematis untuk mengeksplorasi data - memahami struktur, karakteristik, pola dan anomali

### Tujuan EDA:

* Memahami betuk dan struktur data (berapa baris, kolom, tipe data apa saja)
* Mendeteksi masalah tersembunyi yang lolos dari proses cleaning
* Menemukan pola dan hubungan antar variabel
* Membentuk hipotesis yang bisa diuji lebih lanjut
* Menentukan strategi modeling yang tepat (misalnya: apakah perlu transformasi, apakah target imbalance, dll)

> Analogi sederhana: EDA itu seperti seorang dokter yang memeriksa pasien sebelum menentukan diagnosis. Anda tidak langsung meresepkan obat (model) tanpa memeriksa gejala (data) terlebih dahulu.

## 1.2 Urutan Sistematis EDA

EDA punya urutan logis supaya setiap tahap membangun pemahaman untuk tahap berikutnya:

```text
1. Struktur & Verifikasi Data   ← Anda di sini (Modul 01)
2. Univariate Analysis          (satu variabel per waktu)
3. Bivariate Analysis           (hubungan dua variabel)
4. Multivariate Analysis        (hubungan banyak variabel)
5. Anomaly & Class Imbalance
6. Hypothesis Generation

### Kenapa Urutan Ini Penting?

* EDA harus dilakukan secara bertahap dan sistematis, bukan secara acak.
* Bivariate Analysis tidak dapat dilakukan dengan baik jika Anda belum memahami:

  * Tipe data setiap variabel
  * Struktur data
  * Kualitas dan kebersihan data
* Kesalahan pada tahap awal dapat menyebabkan kesimpulan yang salah pada tahap analisis berikutnya.
* Modul 01 — Struktur & Verifikasi Data menjadi gerbang masuk menuju seluruh proses EDA.
* Ketelitian pada Modul 01 akan menentukan kualitas:

  * Univariate Analysis
  * Bivariate Analysis
  * Multivariate Analysis
  * Analisis anomaly dan class imbalance
  * Hypothesis generation
* Intinya: jika fondasi datanya salah, analisis di atasnya juga berisiko salah.

### Pentingnya Verifikasi Struktur Data pada Tahap Awal EDA

* Verifikasi struktur data merupakan tahap fundamental dalam proses Exploratory Data Analysis (EDA).
* Tahap ini bertujuan untuk memperoleh pemahaman awal mengenai karakteristik dataset sebelum dilakukan analisis lebih lanjut.
* Informasi mengenai ukuran dataset perlu diperiksa untuk mengetahui jumlah observasi dan variabel yang tersedia.
* Daftar kolom juga perlu diverifikasi untuk memastikan bahwa variabel yang dianalisis sesuai dengan tujuan penelitian atau analisis.
* Verifikasi tersebut dapat mencegah beberapa permasalahan, seperti:

  * Analisis terhadap variabel yang tidak relevan.
  * Terlewatnya variabel yang memiliki peran penting.
  * Ketidaksesuaian antara variabel yang diharapkan dengan variabel yang tersedia dalam dataset.
* Dengan demikian, pemeriksaan struktur dataset pada tahap awal dapat menjadi dasar yang valid dan sistematis bagi pelaksanaan analisis pada tahap EDA berikutnya.


## 1.4 Tipe Data & Identifikasi Numerical / Categorical / Datetime

Setiap kolom perlu diklasifikasikan ke salah satu kategori berikut, karena jenis analisis yang bisa dilakukan berbeda tergantung tipenya:

### Klasifikasi Tipe Data

1. Numerical — data yang dapat diukur dan digunakan dalam operasi matematis.
   * Continuous — memiliki nilai yang dapat berupa desimal.
     * Contoh: gaji, berat badan, tinggi badan.
   * Discrete — memiliki nilai berupa bilangan cacah atau bilangan bulat.
     * Contoh: jumlah anak, jumlah unit, jumlah transaksi.

2. Categorical — data yang merepresentasikan kategori atau label tertentu.
   * Nominal — kategori tidak memiliki tingkatan atau urutan.
     * Contoh: kota, departemen, jenis kelamin.
   * Ordinal — kategori memiliki tingkatan atau urutan yang bermakna.

     * Contoh: tingkat pendidikan, level jabatan, tingkat kepuasan.

3. Datetime — data yang merepresentasikan tanggal atau waktu.
   * Contoh: tanggal bergabung, tanggal transaksi, timestamp.


### Masalah Umum dalam Tipe Data

* Salah satu masalah yang paling sering ditemukan dalam tahap verifikasi data adalah ketidaksesuaian antara tipe data aktual dan tipe data yang terbaca oleh sistem.
* Kolom tanggal atau numerik dapat terbaca sebagai `object` karena:

  * Format tanggal tidak standar.
  * Adanya simbol mata uang, seperti `Rp` atau `$`.
  * Penggunaan koma sebagai pemisah ribuan.
  * Adanya karakter atau teks tambahan pada nilai numerik.
* Kondisi tersebut perlu diperbaiki sebelum dilakukan analisis statistik.
* Jika tidak dikoreksi sejak awal, masalah tipe data dapat menyebabkan:

  * Perhitungan statistik menghasilkan nilai yang tidak valid.
  * Operasi matematis gagal atau menghasilkan error.
  * Analisis pada tahap EDA berikutnya menjadi tidak akurat.
* Oleh karena itu, verifikasi dan koreksi tipe data merupakan bagian penting dalam tahap awal EDA untuk memastikan data siap digunakan dalam analisis selanjutnya.


## 1.5 Verifikasi Hasil Cleaning Fase 1

### EDA vs Data Cleaning

* Data cleaning bertujuan memperbaiki masalah pada data, seperti missing value, duplikasi, format tanggal, dan tipe data.
* EDA dilakukan setelah cleaning untuk memahami dan memverifikasi bahwa hasil cleaning sudah benar.
* EDA tetap dapat menemukan masalah baru yang sebelumnya belum terdeteksi.
* Jika ditemukan masalah, data dapat dikembalikan ke tahap cleaning untuk diperbaiki.

> Intinya: Data cleaning memperbaiki data, sedangkan EDA memverifikasi dan memahami data yang telah dibersihkan.


Pertanyaan yang harus dijawab di tahap ini:

* Apakah missing value yang seharusnya sudah ditangani, benar-benar sudah 0 (atau sesuai treatment yang direncanakan)?
* Apakah duplicate yang seharusnya dihapus, benar-benar sudah hilang?
* Apakah tipe data sudah sesuai ekspektasi setelah proses cleaning?

## 1.6 Cek Missing Value & Duplicate

* Missing value — nilai yang kosong/hilang pada suatu sel. Dicek dengan menghitung jumlah null/NaN per kolom.
* Duplicate — baris yang identik atau merepresentasikan entitas yang sama lebih dari sekali (misal employee_id yang sama muncul dua kali).

## 1.7 Membedakan EDA dengan Data Cleaning

| Aspek     | Data Cleaning                                                            | EDA                                                                  |
| --------- | ------------------------------------------------------------------------ | -------------------------------------------------------------------- |
| Tujuan    | Memperbaiki data yang rusak atau tidak sesuai                            | Memahami karakteristik dan pola data                                 |
| Aktivitas | Mengisi/menghapus missing value, menghapus duplikasi, memperbaiki format | Menghitung statistik, membuat visualisasi, mencari pola dan hubungan |
| Sifat     | Mengubah data                                                            | Mengamati dan menganalisis data                                      |
| Waktu     | Sebelum EDA (Fase 1)                                                     | Setelah data dibersihkan (Fase 2)                                    |


### Penanganan Masalah yang Ditemukan Saat EDA

* EDA dapat menemukan masalah yang sebelumnya belum terdeteksi saat data cleaning.
* Contohnya adalah outlier ekstrem yang terlihat mencurigakan.
* Masalah tersebut sebaiknya dicatat dan didokumentasikan sebagai temuan EDA.
* Jangan langsung menghapus atau mengubah data hanya karena terlihat sebagai outlier.
* Outlier perlu dianalisis lebih lanjut untuk menentukan apakah:

  * Merupakan kesalahan data.
  * Merupakan kejadian yang valid.
  * Memiliki informasi penting bagi analisis.
* Keputusan mengenai tindakan terhadap outlier dilakukan setelah analisis lebih mendalam pada tahap anomaly analysis.

> Intinya: temukan → dokumentasikan → analisis → baru tentukan tindakan.


## 1.8 Membuat Initial Data Profile

* Setelah seluruh pemeriksaan awal selesai, hasilnya dirangkum dalam sebuah Initial Data Profile.
* Initial Data Profile berfungsi sebagai “kartu identitas” dataset yang memberikan gambaran kondisi awal data.
* Informasi yang umumnya dicantumkan meliputi:

  * Jumlah baris dan kolom
  * Tipe data setiap kolom
  * Jumlah missing value
  * Jumlah data duplikat
  * Ringkasan statistik dasar
* Profil ini menjadi titik acuan untuk membandingkan dan mengevaluasi kondisi data selama proses EDA berikutnya.
* Dengan adanya Initial Data Profile, perubahan atau temuan baru pada tahap EDA dapat ditelusuri dengan lebih sistematis.

# Bagian 3 — Latihan

Kerjakan latihan berikut menggunakan dataset Anda sendiri (boleh dataset hasil cleaning dari project sebelumnya, atau dataset publik apa pun yang ingin Anda pelajari).

````python
import pandas as pd
import numpy as np

# ------------------------------------------------------------------
# Dataset sintetis: Data Karyawan Perusahaan Manufaktur
# Anggap ini adalah output Fase 1 (cleaning) yang akan Anda VERIFIKASI
# di Modul 01 — bukan dataset yang sudah pasti 100% bersih.
# ------------------------------------------------------------------

np.random.seed(42)  # supaya hasil random selalu sama tiap dijalankan

n = 200  # jumlah baris

# --- Kolom identifier ---
employee_id = list(range(1001, 1001 + n))

# --- Kolom numerical ---
usia = np.random.randint(21, 58, size=n).astype(float)
gaji = np.random.randint(4_000_000, 15_000_000, size=n)
lama_bekerja_bulan = np.random.randint(1, 120, size=n)

# --- Kolom categorical ---
departemen = np.random.choice(
    ['Produksi', 'Quality Control', 'Logistik', 'HR', 'Finance'],
    size=n, p=[0.4, 0.2, 0.2, 0.1, 0.1]  # sengaja tidak merata → cardinality realistis
)
status_karyawan = np.random.choice(['Aktif', 'Resign'], size=n, p=[0.85, 0.15])
level_pendidikan = np.random.choice(
    ['SMA', 'D3', 'S1', 'S2'], size=n, p=[0.3, 0.2, 0.4, 0.1]
)

# --- Kolom datetime, SENGAJA dibuat sebagai string dengan format campuran ---
tanggal_masuk = pd.date_range('2015-01-01', '2024-01-01', periods=n).strftime('%Y-%m-%d').tolist()

# --- Gaji, SENGAJA diberi format string dengan simbol supaya jadi 'object' ---
gaji_str = ['Rp' + format(g, ',').replace(',', '.') for g in gaji]

df = pd.DataFrame({
    'employee_id': employee_id,
    'usia': usia,
    'departemen': departemen,
    'level_pendidikan': level_pendidikan,
    'status_karyawan': status_karyawan,
    'gaji': gaji_str,                     # jebakan: harus numeric, masih string
    'lama_bekerja_bulan': lama_bekerja_bulan,
    'tanggal_masuk': tanggal_masuk         # jebakan: harus datetime, masih string
})

# ------------------------------------------------------------------
# Sisipkan missing value secara acak (simulasi data belum sepenuhnya bersih)
# ------------------------------------------------------------------
missing_idx_usia = np.random.choice(df.index, size=8, replace=False)
df.loc[missing_idx_usia, 'usia'] = np.nan

missing_idx_pendidikan = np.random.choice(df.index, size=5, replace=False)
df.loc[missing_idx_pendidikan, 'level_pendidikan'] = np.nan

# ------------------------------------------------------------------
# Sisipkan baris duplicate (simulasi input ganda / data entry error)
# ------------------------------------------------------------------
df = pd.concat([df, df.iloc[[10, 25, 47]]], ignore_index=True)

# ------------------------------------------------------------------
# Acak urutan baris supaya duplicate tidak langsung terlihat berurutan
# ------------------------------------------------------------------
df = df.sample(frac=1, random_state=1).reset_index(drop=True)

print("Dataset siap. Shape:", df.shape)
df.head()
````
