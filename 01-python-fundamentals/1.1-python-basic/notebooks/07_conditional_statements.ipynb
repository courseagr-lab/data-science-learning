# 1. Basic Conditional

## `if`

* Blok kode di dalam if hanya dijalankan kalau kondisinya bernilai True
* Kondisi bisa berupa hasil perbandingan (>, <, ==, dst — dari materi 02), atau nilai apa pun yang dicek truthy/falsy-nya (juga dari materi 02)
* Python wajib pakai indentasi (biasanya 4 spasi) untuk menandai blok kode mana yang termasuk bagian if — ini bukan gaya penulisan opsional, tapi aturan sintaks

````python
umur = 20

if umur >= 20:
    print("Sudah dewasa")
````

## `elif`

* Singkatan dari "else if" — dipakai kalau ada lebih dari satu kemungkinan kondisi yang perlu dicek secara berurutan
* Python akan mengecek dari atas ke bawah, dan berhenti di kondisi pertama yang  True — kondisi setelahnya tidak akan dicek lagi walau sebenarnya juga True

````python
nilai = 75

if nilai >= 90:
    print("A")
elif nilai >= 80:
    print("B")
elif nilai >= 70:
    print("C")          # ini yang tercetak, karena 75 masuk kondisi ini
elif nilai >= 60:
    print("D")           # TIDAK dicek lagi, walau 75 >= 60 juga True
else:
    print("E")
````

## `else`

* Blok "penampung" untuk kondisi yang tidak cocok dengan semua if/elif di atasnya
* Sifatnya opsional — kamu boleh menulis if tanpa else sama sekali
* Tidak butuh kondisi apa pun, karena maknanya "kalau semua di atas tidak ada yang benar"

````python
status_pembayaran = "gagal"

if status_pembayaran == "sukses":
    print("Transaksi berhasil")
else:
    print("Transaksi bermasalah")   # ini yang tercetak
````

> Poin penting: if bisa berdiri sendiri, elif dan else selalu mengikuti if — tidak bisa elif/else tanpa if di atasnya.

# 2. Boolean Conditions

## Comparison

* Operator perbandingan (==, !=, >, <, >=, <=) — sudah kamu pelajari di materi 02 — adalah bahan dasar paling umum untuk kondisi if
* Hasil dari operator ini selalu True atau False, itulah yang dibaca if

````python
harga = 50000
budget = 40000

if harga > budget:
    print("Melebihi budget")
````

## Logical operators

* and → kondisi gabungan bernilai True hanya jika semua bagian bernilai True
* or → kondisi gabungan bernilai True jika salah satu bagian bernilai True
* not → membalik nilai boolean (True jadi False, dan sebaliknya)
* Ketiganya sudah kamu pelajari di materi 02 — di sini kita pakai langsung untuk kondisi if yang lebih kompleks

````python
umur = 25
punya_ktp = True

if umur >= 17 and punya_ktp:
    print("Boleh membuat SIM")

is_weekend = False
is_libur_nasional = True

if is_weekend or is_libur_nasional:
    print("Hari ini libur")

sedang_maintenance = False
if not sedang_maintenance:
    print("Sistem bisa diakses")
````

## Multiple conditions

* Kondisi majemuk bisa digabung lebih dari dua, dan bisa dikombinasikan dengan kurung () untuk mengatur urutan evaluasi — mirip precedence pada operator aritmatika (materi 02)
* Tanpa kurung, and dievaluasi lebih dulu dibanding or (mirip perkalian lebih dulu dari penjumlahan)

````python
usia = 22
punya_tiket = True
sudah_vaksin = True

# Tanpa kurung -> and dievaluasi duluan, bisa ambigu maksudnya
if usia >= 18 and punya_tiket or sudah_vaksin:
    print("Boleh Masuk")
    
# Dengan kurung -> maksudnya jadi jelas, tidak perlu menebak precedence
if usia >= 18 and (punya_tiket or sudah_vaksin):
    print("Boleh masuk (lebih jelas maksudnya)")
````

> Kesalahan umum pemula: menggabungkan banyak and/or tanpa kurung, lalu bingung sendiri kenapa hasilnya tidak sesuai ekspektasi. Kalau kondisimu sudah lebih dari 2 bagian, selalu pertimbangkan pakai kurung eksplisit — sama seperti prinsip operator precedence di materi 02.

# 3. Nested Conditions

## Nested if

* Nested condition artinya ada if di dalam blok if lainnya
* Berguna saat kondisi kedua hanya relevan dicek setelah kondisi pertama terpenuhi

````python
umur = 25
punya_ktp = True

if umur >= 17:
    if punya_ktp:
        print("Boleh membuat SIM")
    else:
        print("Perlu KTP dulu")
else:
    print("Belum cukup umur")
````

## Kapan diperlukan

* Dipakai saat kondisi kedua secara logis memang hanya masuk akal dicek setelah kondisi pertama benar — bukan sekadar dua kondisi independen
* Contoh masuk akal: cek "apakah user login" dulu, baru di dalamnya cek "apakah dia admin" — karena mengecek status admin tidak relevan kalau user belum login sama sekali

````python
sudah_login = True
role = "admin"

if sudah_login:
    if role == "admin":
        print("Akses penuh diberikan")
    else:
        print("Akses terbatas")
else:
    print("Silakan login dulu")
````

## Menghindari nesting berlebihan

* Nested condition yang bertumpuk terlalu dalam (3-4 level ke bawah) membuat kode sulit dibaca — ini disebut istilah informal "arrow code" karena bentuknya makin menjorok ke kanan seperti anak panah
* Kalau kondisinya sebenarnya independen (tidak saling bergantung), lebih baik pakai and daripada nested if

````python
# KURANG BAIK -> nested padahal sebenarnya bisa digabung
umur = 25
punya_ktp = True
punya_sim_lama = False

if umur >= 17:
    if punya_ktp:
        if not punya_sim_lama:
            print("Bisa mengajukan SIM baru")

# LEBIH BAIK -> digabung jadi satu kondisi dengan and
if umur >= 17 and punya_ktp and not punya_sim_lama:
    print("Bisa mengajukan SIM baru")
````

Kalau kondisinya banyak kemungkinan sejajar (bukan bertingkat), pakai elif biasa, bukan nested if-else berulang

````python
nilai = 75

# KURANG BAIK -> nested if-else berulang, susah dibaca
if nilai >= 90:
    print("A")
else:
    if nilai >= 80:
        print("B")
    else:
        if nilai >= 70:
            print("C")
        else:
            print("D")

# LEBIH BAIK -> pakai elif, lebih rata dan mudah dibaca
if nilai >= 90:
    print("A")
elif nilai >= 80:
    print("B")
elif nilai >= 70:
    print("C")
else:
    print("D")
````

> Prinsip praktis: kalau kamu menemukan dirimu menulis nested if lebih dari 2 level, berhenti sejenak dan tanya — apakah ini bisa disederhanakan jadi and/or, atau jadi elif sejajar? Kode yang rapi biasanya "lebar" (banyak elif sejajar), bukan "dalam" (banyak nested).

# 4. Conditional Expression

## Ternary expression

* Cara menulis if-else sederhana dalam satu baris, biasanya untuk sekadar memilih salah satu dari dua nilai
* Bukan pengganti if biasa untuk logika kompleks — ini murni soal keringkasan untuk kasus simpel

````python
# Bentuk if-else biasa
umur = 20
if umur >= 18:
    status = "Dewasa"
else:
    status = "Anak-anak"

print(status)   # Dewasa

# Bentuk ternary -> satu baris, hasil yang sama persis
status2 = "Dewasa" if umur >= 18 else "Anak-anak"
print(status2)   # Dewasa
````

Pola umum: nilai_jika_true if kondisi else nilai_jika_false

````python
nilai = 55
keterangan = "Lulus" if nilai >= 60 else "Tidak Lulus"
print(keterangan)   # Tidak Lulus

# Bisa langsung dipakai di dalam f-string juga (dari materi 02)
harga = 15000
diskon_persen = 20 if harga > 10000 else 0
print(f"Diskon: {diskon_persen}%")   # Diskon: 20%
````

## Kapan digunakan

* Cocok untuk kondisi sederhana — cuma memilih satu dari dua nilai, tanpa logika tambahan
* Hindari dipakai untuk kondisi kompleks atau nested — itu justru bikin kode lebih susah dibaca, bukan lebih ringkas

````python
# COCOK -> sederhana, satu keputusan, satu baris cukup jelas
minimum_pembelian = 50000
total_belanja = 75000
status_gratis_ongkir= "Gratis Ongkir" if total_belanja >= minimum_pembelian else "Tidak Gratis Ongkir"
print(status_gratis_ongkir)

# KURANG COCOK -> dipaksa jadi ternary padahal logikanya kompleks, malah susah dibaca
# nilai_huruf = "A" if nilai >= 90 else "B" if nilai >= 80 else "C" if nilai >= 70 else "D"
# -> untuk kasus begini, if/elif/else biasa jauh lebih jelas dibaca
````

> Kesalahan umum pemula: memaksakan ternary bertingkat (ternary di dalam ternary) supaya kode "kelihatan singkat". Ini justru kontraproduktif — tujuan ternary adalah keterbacaan untuk kasus simpel, bukan meminimalkan jumlah baris dengan mengorbankan kejelasan.

3. Status transaksi

    Diberikan tiga variabel: saldo, harga_barang, dan stok_tersedia (boolean). Buat logika:

    * Kalau stok_tersedia bernilai False → cetak "Barang habis"
    * Kalau stok tersedia tapi saldo < harga_barang → cetak "Saldo tidak cukup"
    * Kalau stok tersedia dan saldo cukup → cetak "Transaksi berhasil", lalu hitung dan cetak sisa saldo

    Coba kerjakan dengan nested if dulu, lalu coba versi keduanya pakai kondisi gabungan (and) — bandingkan mana yang menurutmu lebih mudah dibaca untuk kasus ini.

````python
saldo = 1000
harga_barang = 200
stok_tersedia = False

if stok_tersedia:
    if saldo >= harga_barang:
        print("Transaksi Berhail")
        
        sisa_saldo = saldo - harga_barang
        print(f"Sisa Saldo {sisa_saldo}")
    else:
        print("Saldo tidak cukup")
else:
    print("Barang Habis")
````

4. Validasi data sederhana
    Diberikan dictionary (dari materi 05) seperti berikut:

    ```python
    python
    user_input = {"nama": "Andi", "umur": 25, "email": ""}
    ```

    Buat validasi yang mengecek:

    * Apakah "nama" terisi (bukan string kosong) — manfaatkan konsep truthy/falsy dari materi 02
    * Apakah "umur" bernilai masuk akal (misal antara 1-120)
    * Apakah "email" terisi — kalau kosong, cetak pesan peringatan, tapi program tetap lanjut (bukan berhenti total)

    Gabungkan ketiga pengecekan itu, lalu cetak "Data valid" hanya kalau nama dan umur valid (email boleh kosong tapi tetap kasih peringatan terpisah). Gunakan minimal satu ternary expression di salah satu bagian kode ini.

````python
user_input = {"nama": "Andi", "umur": 25, "email": ""}

# 1. Pengecekan Nama (menggunakan konsep truthy/falsy)
nama_valid = bool(user_input.get("nama"))

# 2. Pengecekan Umur (range 1-120)
umur = user_input.get("umur", 0)
umur_valid = 1 <= umur <= 120

# 3. Pengecekan Email menggunakan Ternary Expression
# Mencetak peringatan langsung jika email kosong, tanpa menghentikan program
email_status = (
    "Email terisi."
    if user_input.get("email")
    else "Peringatan: Email kosong!"
)
print(email_status)

# 4. Validasi Akhir (Nama dan Umur harus valid)
if nama_valid and umur_valid:
    print("Data valid")
else:
    print("Data tidak valid")
````
