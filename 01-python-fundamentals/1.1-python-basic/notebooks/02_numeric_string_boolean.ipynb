# 1. Numeric Data Types

Python punya dua tipe angka utama:
* int - bilangan bulat, tanpa koma (1, 5, 2)
* float - bilangan desimal / bilangan pecahan (50.3, 4.2)

Python menyimpan keduanya dengan cara yang berbeda secara internal, dan ini berpengaruh ke hasil operasi

````python
umur = 25            # int -> bilangan bulat
tinggi_cm = 165.5    # float -> punya desimal

print(type(umur))        # <class 'int'>
print(type(tinggi_cm))   # <class 'float'>

# Perhatikan: walaupun nilainya "bulat", kalau ditulis pakai titik, tetap float
harga = 10000.0
print(type(harga))       # <class 'float'>

# int bisa diubah ke float dan sebaliknya
print(float(umur))       # 25.0
print(int(tinggi_cm))    # 165 -> desimal dibuang, BUKAN dibulatkan
````

> Kenapa penting di data science: saat baca data dari CSV, kolom yang seharusnya bilangan bulat (misal jumlah_anak) tapi terbaca sebagai float sering menandakan ada nilai kosong (NaN) di kolom itu

# 2. Arithmetic Operators

Operasi dasar matematika

| Operator | Arti                            | Contoh   | Hasil      |
| -------- | ------------------------------- | -------- | ---------- |
| `+`      | Penjumlahan                     | `5 + 3`  | `8`        |
| `-`      | Pengurangan                     | `5 - 3`  | `2`        |
| `*`      | Perkalian                       | `5 * 3`  | `15`       |
| `/`      | Pembagian (hasil selalu float)  | `5 / 3`  | `1.666...` |
| `//`     | Floor division (bulat ke bawah) | `5 // 3` | `1`        |
| `%`      | Modulus (sisa bagi)             | `5 % 3`  | `2`        |
| `**`     | Pangkat                         | `5 ** 3` | `125`      |


````python
a = 17
b = 5

print(a + b)   # 22
print(a - b)   # 12
print(a * b)   # 85
print(a / b)   # 3.4  -> pembagian biasa
print(a // b)  # 3    -> hanya bagian bulatnya
print(a % b)   # 2    -> sisa dari 17 dibagi 5
print(a ** b)  # 1419857 -> 17 pangkat 5
````

# 3. Operator Precedence (Urutan Operasi)

Aturannya mirip matematika sekolah: kurung → pangkat → kali/bagi → tambah/kurang, dan kalau levelnya sama, dieksekusi dari kiri ke kanan (kecuali pangkat, yang dari kanan ke kiri).

````python
hasil1 = 2 + 3 * 4       # 14, bukan 20 -> perkalian dikerjakan duluan
hasil2 = (2 + 3) * 4     # 20 -> kurung "memaksa" dikerjakan lebih dulu
hasil3 = 2 ** 3 ** 2     # 512 -> dievaluasi kanan dulu: 3**2=9, lalu 2**9=512

print(hasil1, hasil2, hasil3)
````

> Prinsip praktis: kalau ekspresi kamu punya lebih dari 2 operator berbeda, sebaiknya tetap pakai kurung eksplisit — bukan karena Python akan salah, tapi supaya orang lain (dan kamu sendiri beberapa bulan lagi) tidak perlu menebak maksudnya.

# 4. / vs //

Hitung rata rata atau presentase:
* / (True division): selalu menghasilkan float, presisi penuh
* // (Floor division): hasil dibulatkan kebawah

````python
print(9 / 2)     # 4.5   -> true division
print(9 // 2)    # 4     -> floor division, buang desimalnya

# Hati-hati dengan angka negatif! "ke bawah" berarti ke arah negatif tak hingga
print(-9 // 2)   # -5, BUKAN -4

# Contoh penggunaan nyata: membagi rata peserta ke dalam grup
total_peserta = 23
kapasitas_per_grup = 5

jumlah_grup_penuh = total_peserta // kapasitas_per_grup   # 4 grup penuh
sisa_peserta = total_peserta % kapasitas_per_grup          # 3 peserta sisa

print(f"{jumlah_grup_penuh} grup penuh, sisa {sisa_peserta} peserta")
````

# 5. Floating Point Precision

````python
print(0.1 + 0.2)          # 0.30000000000000004 -> bukan 0.3!
print(0.1 + 0.2 == 0.3)   # False
````

* Komputer menyimpan angka desimal dalam sistem biner.
* Tidak semua angka desimal memiliki representasi biner yang eksak.
* Contohnya:

  * `1/3` dalam desimal → `0.333...` tanpa akhir.
  * `0.1` dalam biner → juga tidak dapat direpresentasikan secara eksak.
* Akibatnya, komputer menyimpan nilai pendekatan dari angka tersebut.
* Terdapat selisih pembulatan (rounding error) yang sangat kecil dan biasanya tidak terlihat.
* Inilah salah satu alasan mengapa operasi seperti `0.1 + 0.2` dapat menghasilkan `0.30000000000000004` di Python.


````python
# Solusi 1: round() untuk membulatkan hasil tampilan
print(round(0.1 + 0.2, 2))    # 0.3

# Solusi 2: kalau perlu membandingkan float, jangan pakai == langsung.
# Gunakan toleransi kecil (epsilon)
a = 0.1 + 0.2
b = 0.3
selisih = abs(a - b)
print(selisih < 1e-9)   # True -> ini cara aman membandingkan float
````

> Kenapa penting: saat validasi data (misal cek total persentase = 100%), kalau kamu pakai == langsung, validasi bisa gagal padahal datanya sebenarnya sudah benar — hanya karena pembulatan floating point.

Setiap karakter dalam string punya alamat(index), dimulai dari 0

## Slicing

* Mengambil sebagian string dengan format `[mulai:akhir:step]`.
* `mulai` → indeks awal, termasuk dalam hasil.
* `akhir` → indeks akhir, tidak termasuk dalam hasil (eksklusif).
* `step` → menentukan jarak/lompatan antar karakter.
* Batas `akhir` yang eksklusif dapat menyebabkan bug “off-by-one” yang sering membingungkan pemula.


````python
kata = "Python"

print(kata[0:3])    # 'Pyt' -> index 0,1,2 (index 3 TIDAK ikut)
print(kata[2:])      # 'thon' -> dari index 2 sampai akhir
print(kata[:3])      # 'Pyt' -> dari awal sampai sebelum index 3
print(kata[::-1])    # 'nohtyP' -> step -1 = dibalik
print(kata[::2])     # 'Pto' -> ambil tiap 2 karakter
````

# String Methods

````python
teks = "  Halo Dunia Data Science  "

# .strip() -> menghapus spasi (atau karakter tertentu) di awal & akhir
print(teks.strip())            # "Halo Dunia Data Science"

# .split() -> memecah string jadi list, berdasarkan pemisah (default: spasi)
kalimat = "belajar python itu seru"
print(kalimat.split())         # ['belajar', 'python', 'itu', 'seru']

csv_row = "Andi,25,Jakarta"
print(csv_row.split(","))      # ['Andi', '25', 'Jakarta']

# .join() -> kebalikan dari split, menggabungkan list jadi satu string
kata_list = ['belajar', 'python', 'itu', 'seru']
print(" ".join(kata_list))     # "belajar python itu seru"
print("-".join(kata_list))     # "belajar-python-itu-seru"

# .replace() -> mengganti substring tertentu
kalimat2 = "saya suka kucing"
print(kalimat2.replace("kucing", "anjing"))   # "saya suka anjing"

# .lower() dan .upper() -> ubah ke huruf kecil/besar semua
print("Data Science".lower())   # "data science"
print("Data Science".upper())   # "DATA SCIENCE"

# .title() -> huruf awal tiap kata jadi kapital
print("machine learning engineer".title())   # "Machine Learning Engineer"
````

> Semua method di atas tidak mengubah string asli (karena string itu immutable) — mereka mengembalikan string baru. Jadi selalu simpan hasilnya ke variabel kalau mau dipakai lagi.

# 8. F-String

F-string adalah cara untuk menggabungkan teks dengan variabel

````python
nama = "Andi"
umur = 25

# Basic f-string -> taruh 'f' sebelum tanda kutip
print(f"Nama saya {nama}, umur {umur} tahun")
# "Nama saya Andi, umur 25 tahun"

# Bisa juga langsung berisi ekspresi/perhitungan di dalam { }
harga = 15000
jumlah = 3
print(f"Total belanja: {harga * jumlah}")   # "Total belanja: 45000"

# Formatting angka -> misal untuk 2 angka di belakang koma
pi = 3.14159265
print(f"Nilai pi: {pi:.2f}")    # "Nilai pi: 3.14"

# Formatting dengan pemisah ribuan
total = 1500000
print(f"Total: Rp{total:,}")    # "Total: Rp1,500,000"
````

# 9. Boolean

boolean cuma punya dua nilai: True dan False

````python
# Comparison operators
print(5 > 3)     # True
print(5 < 3)     # False
print(5 == 5)    # True  -> sama dengan
print(5 != 3)    # True  -> tidak sama dengan
print(5 >= 5)    # True  -> lebih besar atau sama dengan
print(5 <= 4)    # False

# and -> True hanya jika KEDUA kondisi True
umur = 20
punya_ktp = True
print(umur >= 17 and punya_ktp)   # True

# or -> True jika SALAH SATU kondisi True
punya_kartu_pelajar = False
punya_kartu_mahasiswa = True
print(punya_kartu_pelajar or punya_kartu_mahasiswa)   # True

# not -> membalik nilai boolean
sedang_hujan = False
print(not sedang_hujan)   # True
````

# 10. Truthy/Falsy

* Setiap nilai di Python dapat dievaluasi sebagai `True` atau `False` dalam konteks boolean.
* Nilai yang dianggap `True` disebut truthy.
* Nilai yang dianggap `False` disebut falsy.
* Nilai truthy/falsy tidak harus berupa `True` atau `False` secara eksplisit.
* Konsep ini sering digunakan dalam:

  * `if`
  * `while`
  * Validasi kondisi
  * Pengecekan apakah suatu data tersedia atau kosong
* Contoh nilai yang umumnya falsy:

  * `False`
  * `None`
  * `0`
  * `0.0`
  * `""` (string kosong)
  * `[]` (list kosong)
  * `{}` (dictionary kosong)
* Sebagian besar nilai lainnya dianggap truthy.
* Memahami Truthy/Falsy penting karena kesalahan dalam evaluasi nilai kosong atau `0` dapat menyebabkan bug tersembunyi, terutama saat melakukan validasi data.


````python
print(bool(0))       # False -> angka nol
print(bool(''))      # False -> string kosong
print(bool([]))      # False -> list kosong
print(bool({}))      # False -> dictionary kosong
print(bool(None))    # False -> None (tidak ada nilai)

# Selain itu, HAMPIR SEMUA nilai lain dianggap Truthy (True)
print(bool(1))        # True
print(bool(-1))       # True -> bahkan angka negatif tetap Truthy!
print(bool("0"))      # True -> string "0" itu Truthy, beda dengan angka 0!
print(bool([0]))      # True -> list berisi satu elemen, walau elemennya 0
````

Contoh kenapa ini penting dalam praktik nyata:

````python
nama_input = ""

# Cara yang RENTAN bug -> mengira ini mengecek "apakah nama None"
if nama_input:
    print("Nama diisi")
else:
    print("Nama kosong")   # ini yang tercetak, karena '' itu Falsy

# Jebakan klasik: 0 dianggap "tidak ada data", padahal 0 itu data yang valid
saldo = 0
if saldo:
    print("Ada saldo")
else:
    print("Saldo kosong / tidak terdeteksi")   # ini keliru! saldo = 0 itu valid, bukan "kosong"

# Cara yang lebih AMAN dan eksplisit untuk kasus di atas
if saldo is not None:
    print(f"Saldo Anda: {saldo}")   # akan mencetak "Saldo Anda: 0" dengan benar
````

> Kesalahan umum pemula: menganggap 0, '', None itu semuanya "sama-sama kosong". Padahal secara bisnis, 0 bisa jadi nilai yang valid (misal saldo Rp0, bukan berarti datanya hilang). Selalu tanyakan: saya mau cek "apakah ada isinya" atau "apakah nilainya bukan None"? Dua hal ini beda.
