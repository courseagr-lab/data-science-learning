# String

* String: squence of caracters.
* Kumpulan karakter yang bisa diakses berdasarkan posisi
* String bersifat immutable > Setelah string dibuat, karakternya tidak bisa diubah satu persatu
* Setiap operasi yang "mengubah" string sebenarnya membuat string baru di memori. 

```python
name = "Python"
name = "J" + name[1:]
print(name)
```

Why immutable important for DS? :
* Karena saat melakukan transformasi string kolom pandas - lower,strip,replacr - tidak mengubah string yang ada, tetapi membuat string baru untuk setiap baris
* berdampak pada memori dan performa saat dataset besar.

Python melakukan string intering untuk setiap string pendek yang terlihat seperti identifier
> Artinya dua variabel dengan nilai string yang sama bisa menunjuk ke objek yang sama di memori

```python
a = "hello"
b = "hello"
c = "hello world"
d = "hello world"

print(a is b)   # True  → Python intern string pendek ini
print(c is d)   # False → tidak selalu di-intern untuk string panjang
print(a == b)   # True  → selalu gunakan == untuk perbandingan nilai
```

selalu gunakan == untuk membandingkan nilai string, jangan is.

# Cara Membuat String

```python
# Single quote
s1 = 'Hello'

# Double quote
s2 = "Hello"

# Keduanya identik — pilih satu dan konsisten
print(s1 == s2)   # True

# Kapan single quote lebih nyaman
s3 = 'He said "hello" to her'

# Kapan double quote lebih nyaman
s4 = "It's a great day"

# Triple quote — untuk multiline
s5 = """
Ini baris pertama
Ini baris kedua
Ini baris ketiga
"""

s6 = '''Juga bisa
dengan single quote
triple'''

print(s5)
print(s6)
```

# Escape Characters

```python
# Newline
print("Baris 1\nBaris 2")

# Tab
print("Nama:\tAndi")

# Backslash itu sendiri
print("C:\\Users\\Andi\\Documents")

# Single quote di dalam single-quoted string
print('It\'s fine')

# Double quote di dalam double-quoted string
print("He said \"hello\"")

# Unicode character
print("\u00e9")    # é
print("\u2605")    # ★
```

Yang sering relevan di data science: path file di Windows menggunakan backslash. Cara yang lebih aman adalah menggunakan raw string.

```python
text = "Python"
#       012345   → index positif
#      -6-5-4-3-2-1  → index negatif

print(text[0])    # P
print(text[1])    # y
print(text[-1])   # n  → karakter terakhir
print(text[-2])   # o  → karakter kedua dari belakang
```

index negatif menghitung dari belakang. -1 selalu menunjuk ke karakter terakhir, tanpa perlu tahu panjang stringnya.

```python
# Mendapatkan karakter terakhir tanpa tahu panjangnya
filename = "data_2024_cleaned.csv"
print(filename[-4:])   # .csv  → ekstensi file
```

# Slicing - Mengambil substring

```python
text = "Python Data Science"
#       0123456789...

# Sintaks dasar: text[start:stop]
# stop tidak termasuk dalam hasil

print(text[0:6])
print(text[7:11])
print(text[12:19])
print(text[:6])
```

sliccing punya parameter ketiga: step

```python
text = "Python"

print(text[::2])    # Pto - setiap 2 karakter
print(text[::1])    # Python - setiap karakter
print(text[::-1])   # nohtyP - membalikkan string
```

Slccing tidak pernah eror karena out of range

```python
text = "Python"

print(text[0:100])   # Python  → tidak error, otomatis dibatasi
print(text[10:20])   # ""      → string kosong, bukan error
```
