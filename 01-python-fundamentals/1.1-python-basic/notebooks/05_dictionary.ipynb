## Key

* key: "label" atau "nama" dari tiap data, harus unik dalam satu dict
* key harus bertipe data immutable - biasanya str, int, atau tuple (list tidak bisa jadi key
* Kalau kamu menulis key yang sama dua kali, value yang terakhir yang akan dipakai (menimpa yang sebelumnya)

Key duplikat -> value terakhir yang menang, tanpa peringatan/error

````python
data = {
    "nama" : "bagus",
    "nama" : "first"
}
print(data)
````

key bisa berupa angka dan tuple

````python
campuran = {
    1: "satu",
    "dua": 2,
    (0, 0): "origin"
}
print(campuran[1])         # "satu"
print(campuran["dua"])     # 2
print(campuran[(0, 0)])    # "origin"
````

# 2. Mengakses Data

## dict(key)

* Mengakses data secara langsung
* key tidak ditemukan = eror > KeyError

````python
siswa = {"nama": "Andi", "umur": 20, "kota": "Jakarta"}

print(siswa["nama"])
print(siswa["Rumah"])
````

## .get()

* Cara mengambil value 
* Jika key tidak ditemukan tidak eror > mengembalikan None

````python
siswa = {"nama": "Andi", "umur": 20, "kota": "Jakarta"}

print(siswa.get("nama"))
````

````python
print(siswa.get("Sekoal"))
````

Bisa berikan value sendiri jika key tidak ditemukan

````python
print(siswa.get("Sekolah", "Tidak diketahui"))
print(siswa.get("Umur", 18))
````

## [] vs .get()

| Aspek               | `dict[key]`                                                                                      | `dict.get(key)`                                                                        |
| ------------------- | ------------------------------------------------------------------------------------------------ | -------------------------------------------------------------------------------------- |
| Key tidak ditemukan | Error (`KeyError`)                                                                               | Kembalikan `None` / default                                                            |
| Kapan dipakai       | Kalau kamu yakin key pasti ada dan ingin segera mengetahui jika ternyata tidak ada (*fail fast*) | Kalau key mungkin tidak ada dan kamu ingin program tetap berjalan dengan nilai default |


````python
# Contoh praktis: data dari API/user input yang field-nya kadang tidak lengkap
data_user = {"nama": "Citra", "umur": 25}   # tidak ada field "email"

# Pakai [] -> berisiko crash program
# email = data_user["email"]   -> KeyError!

# Pakai .get() -> program tetap jalan dengan aman
email = data_user.get("email", "Email tidak tersedia")
print(email)   # 'Email tidak tersedia'
````

> Prinsip praktis: dalam data science, data dari luar (CSV, API, database) sering tidak konsisten strukturnya. Kebiasaan pakai .get() untuk data yang tidak dijamin lengkap akan menyelamatkanmu dari banyak error tak terduga.

# 3. Manipulasi

## Memanipulasi key-value

Cukup tulis key barud dan assign value - jika key tidak ada , otomatis buat pasangan baru

````python
siswa = {"nama": "Andi", "umur": 20}

siswa['Sekolah'] = "Banjr"
print(siswa)
````

## Mengubah value

Sintaks yang sama persis dengan menambah key baru — bedanya, kalau key sudah ada, value-nya di-overwrite (bukan ditambah baru)

````python
siswa['umur'] = 100
print(siswa['umur'])
````

## del

* keywoed untuk menghapus pasangan key, berdasarkan key
* Jika key tidak ditemukan = KeyErros

````python
del siswa['kota']
print(siswa)
````

## .pop()

* Menghapus key tertentu sekaligus mengembalikan valuenya, berdasarkan key
* Bisa diberikan nilai default jika key tidak ditemukan

````python
siswa = {"nama": "Andi", "umur": 21, "kota": "Jakarta"}

kota_dihapus = siswa.pop('kota')
print(kota_dihapus)
print(siswa)
````

````python
# Aman dari error dengan nilai default
sekolah = siswa.pop("sekolah", "Tidak ada data")
print(sekolah)   # 'Tidak ada data' -> tidak error walau key "sekolah" tidak ada
````

## .update()

* Menimpa / menggabungkan dictionary dengan dictionary lain
* Key sudah ada = ditimpa
* Key belum ada = buat baru

````python
siswa = {"nama": "Andi", "umur": 21}

siswa.update({"Kelas": 7,
              "nama": "rEAL"})
print(siswa)
````

Perbedaan del vs .pop():

* del dict[key] → hanya menghapus, tidak mengembalikan apa-apa
* dict.pop(key) → menghapus dan memberi kamu value-nya untuk dipakai lagi, plus bisa aman dengan default

# 4. Dictionary Methods

## .keys()

mengembalikan semua key dalam dictionary (bentuk seperti list, tapi tipe khususnya dict_keys)

````python
siswa = {"nama": "Andi", "umur": 21, "kota": "Jakarta"}

print(siswa.keys())
print(list(siswa.keys())) # ['nama', 'umur', 'kota'] -> ubah jadi list biasa kalau perlu
````

## .values()

Mengembalikan semua value didalam dictioanry

````python
print(siswa.values())          # dict_values(['Andi', 21, 'Jakarta'])
print(list(siswa.values()))    # ['Andi', 21, 'Jakarta']
````

## .items()

* Mengebalian semua pasangan key-value dalam bentuk tuple

````python
print(siswa.items())
# dict_items([('nama', 'Andi'), ('umur', 21), ('kota', 'Jakarta')])

print(list(siswa.items()))
# [('nama', 'Andi'), ('umur', 21), ('kota', 'Jakarta')]
````

# 5. Iterasi Dictionary

## iterasi key

Secara default, for langsung pada dictionary akan mengiterasi key-nya saja

````python
siswa = {"nama": "Andi", "umur": 21, "kota": "Jakarta"}

for k in siswa:
    print(k)
````

Cara yang lebih eksplisit

````python
for k in siswa.keys():
    print(k)
    
````

## Iterasi values

butuh values = .values()

````python
for v in siswa.values():
    print(v)
````

## Iterasi key-value

dapatkan key dan value sekaligus dalam satu perulangan  = .items()

````python
for k, v in siswa.items():
    print(f"Key: {k} dan Value: {v}")
````

> Kenapa .items() paling sering dipakai: karena dalam praktik nyata kamu hampir selalu butuh tahu "key-nya apa" sekaligus "value-nya apa" — misal saat mencetak laporan atau memvalidasi data per field.

````python
# Contoh praktis: mencetak laporan dari dictionary
harga_produk = {"Indomie": 3500, "Beras": 12000, "Telur": 28000}

for produk, harga in harga_produk.items():
    print(f"{produk}: Rp{harga:,}")
# Indomie: Rp3,500
# Beras: Rp12,000
# Telur: Rp28,000
````

# 6. Nested Dictionary

## Dictionary di dalam dictionary

Value dari sebuah dictionary boleh berupa dictionary lain — ini sangat umum untuk merepresentasikan data yang punya struktur berlapis (mirip JSON)

````python
karyawan = {
    "nama": "Andi",
    "alamat": {
        "jalan": "Jl. Merdeka No. 1",
        "kota": "Jakarta",
        "kode_pos": "12345"
    },
    "gaji": {
        "pokok": 5000000,
        "tunjangan": 1000000
    }
}
````

## Mengakses neted data

butuh akses berlapis lapis

````python
print(karyawan["nama"])                    # 'Andi'
print(karyawan["alamat"])                  # {'jalan': ..., 'kota': ..., 'kode_pos': ...}
print(karyawan["alamat"]["kota"])          # 'Jakarta' -> masuk 2 level
print(karyawan["gaji"]["pokok"])           # 5000000

# Mengubah nested value
karyawan["alamat"]["kota"] = "Bandung"
print(karyawan["alamat"]["kota"])   # 'Bandung'

# Menghitung total gaji dari nested dictionary
total_gaji = karyawan["gaji"]["pokok"] + karyawan["gaji"]["tunjangan"]
print(f"Total gaji: Rp{total_gaji:,}")   # Total gaji: Rp6,000,000

# Kombinasi .get() untuk akses nested yang lebih aman
email = karyawan.get("kontak", {}).get("email", "Tidak ada email")
print(email)   # 'Tidak ada email' -> aman walau "kontak" tidak ada sama sekali
````

> Hati-hati: kalau langsung akses nested dengan [] tanpa yakin strukturnya lengkap (misal karyawan["kontak"]["email"] padahal "kontak" tidak ada), akan langsung KeyError. Untuk data yang strukturnya tidak selalu konsisten (umum terjadi di data real-world), kombinasi .get() berlapis seperti contoh di atas jauh lebih aman.

# 7. List of Dictionaries

## Konsep

*  Ini adalah pola paling penting di seluruh materi ini untuk bekal data science: list yang isinya banyak dictionary, di mana tiap dictionary punya struktur key yang sama
* Bayangkan seperti tabel: setiap dictionary = satu baris, setiap key = satu kolom
* Ini persis struktur yang nanti akan kamu ubah jadi Pandas DataFrame

![list_of_dict_structure.png](../assets/list_of_dict_structure.png)g

````python
# Setiap dictionary mewakili satu "baris data" / satu record
data_karyawan = [
    {"nama": "Andi", "umur": 25, "departemen": "IT"},
    {"nama": "Budi", "umur": 30, "departemen": "Finance"},
    {"nama": "Citra", "umur": 28, "departemen": "IT"},
]

print(data_karyawan)
print(len(data_karyawan))
print(type(data_karyawan))
print(type(data_karyawan[0]))
````

## Mengakses field

Akses index list > akses keynya "dapat satu field"

Akses satu record utuh

````python
print(data_karyawan[0])
````

Akses satu field dari satu record

````python
print(data_karyawan[0]['nama'])
print(data_karyawan[1]['umur'])
````

Iterasi Semua record

````python
for karyawan in data_karyawan:
    print(f"{karyawan['nama']} - {karyawan['departemen']}")
````

````python
data_it = []
for k in data_karyawan:
    if k['departemen'] == 'IT':
        data_it.append(k['nama'])

print(data_it)
````

> Kenapa pola ini krusial: hampir semua data API (format JSON) dan hasil query database, kalau diubah ke Python, bentuknya persis seperti ini — list of dictionaries. Saat nanti belajar Pandas, pd.DataFrame(data_karyawan) akan langsung mengubah struktur ini jadi tabel rapi dalam satu baris kode. Jadi kalau pola ini sudah kamu pahami luar-dalam sekarang, transisi ke Pandas nanti akan jauh lebih mulus.
