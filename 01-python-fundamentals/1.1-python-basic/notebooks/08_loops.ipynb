# 1. For Loop

## Konsep Iteration

* Iterasi: mengambil elemen dari sebuah koleksi satu per satu
* iterable — istilah untuk "sesuatu yang bisa diambil elemennya satu per satu"
* `for` loop: bekerja pada iterable > list, string, tuple, set, dan dictionary
* Pola dasar: `for <variabel_sementara> in <iterable>`, lalu blok kode di bawahnya (dengan indentasi, sama seperti if di materi 07) dijalankan sekali untuk tiap elemen

````python
# Tanpa loop -> harus tulis manual satu-satu, tidak scalable
buah = ["apel", "jeruk", "mangga"]
print(buah[0])
print(buah[1])
print(buah[2])
# Bayangkan kalau ada 1000 data -> mustahil ditulis manual seperti ini

print()

# Dengan for loop -> otomatis, berapa pun panjang datanya
for item in buah:
    print(item)
````

> Poin penting: nama variabel setelah for (di contoh ini item) itu bebas tentukan sendiri — Python akan mengisi variabel itu dengan elemen yang sedang diproses di tiap putaran. Beri nama yang deskriptif, bukan sekadar x atau i, supaya kode mudah dibaca

## Iterasi list

Bentuk umum

````python
harga_produk = [15000, 25000, 8000, 32000]

for harga in harga_produk:
    print(f"Rp{harga:,}")   # f-string dari materi 02
````

## Iterasi string

String = iterable > hasilnya iterasi per karakter

````python
kata = "Python"

for huruf in kata:
    print(huruf)
````

## Iterasi Set

Pada set urutan output tidak terjamin

````python
kota_unik = {"Jakarta", "Bandung", "Surabaya"}

for kota in kota_unik:
    print(kota)
# urutan tampilan BISA berbeda-beda tiap kali dijalankan, karena set tidak berurutan
````

# 2. `range()`

range() dipakai saat butuh perulangan berbasis angka, bukan berbasis koleksi data yang sudah ada.

## `range(stop)`

* Menghasilkan urutan angka mulai dari 0, sampai `stop - 1` (batas atas eksklusif, sama seperti slicing)

````python
for i in range(5):
    print(i)
````

## `python range(start, stop)`

* Bisa menentukan awal sendiri

````python
for i in range(2, 6):  # mulai dari dua , berhenti sebelum 6
    print(i)
````

## `range(start, stop, step)`

* Argumen ketiga menentukan lompatan antar angka
* Step bisa negatif untuk menghitung mundur

````python
for i in range(0, 10, 2):
    print(i)
````

````python
for i in range(10, 0, -2):
    print(i)
````

Kapan pakai `range()` dibanding iterasi langsung?

* Gunakan `range()` jika membutuhkan angka/index atau ingin mengulang sebanyak N kali.

  * Contoh: perhitungan, membuat urutan angka, pengulangan tertentu.
* Gunakan iterasi langsung (`for item in list`) jika hanya membutuhkan isi koleksi.

  * Lebih ringkas dan mudah dibaca.
* Prinsip sederhana:

  * Butuh angka/index → `range()`
  * Butuh isi data → iterasi langsung


````python
# KURANG BAIK -> muter-muter pakai range() padahal cuma butuh isinya

buah = ["apel", "jeruk", "mangga"]
for i in range(len(buah)):
    print(i)
````

````python
# LEBIH BAIK -> langsung iterasi, lebih jelas maksudnya
for item in buah:
    print(item)
````

# 3. enumerate()

#### Index + value

* Gunakan `enumerate()` jika membutuhkan index dan value sekaligus.
* Menghasilkan pasangan `(index, value)` yang bisa langsung di-unpack ke dua variabel.
* Lebih ringkas dan umum daripada `for i in range(len(list))`.


````python
buah = ["apel", "jeruk", "mangga"]

for nomor, nama in enumerate(buah):
    print(f"{nomor} buah {nama}")
````

## Kapan menggunakan enumerate()

* Butuh menampilkan nomor urut ke pengguna
* Perlu mengubah elemen di index tertentu sambil tetap iterasi (butuh tahu posisinya)
* Sebagai pengganti pola range(len(list)) yang lebih ringkas dan tidak rawan salah hitung index

````python
# Contoh: mengubah semua harga yang di bawah 10000 jadi tepat 10000
harga = [8000, 15000, 5000, 20000]

for index, nilai in enumerate(harga):
    if nilai < 10000:
        harga[index] = 10000
    
print(harga)
````

* Gunakan `enumerate()` karena perubahan elemen list membutuhkan index.
* `for nilai in harga` hanya mengakses value; mengubah `nilai` tidak mengubah list asli.
* Dengan index: `harga[index] = ...` → elemen list dapat diubah.


# 4. Dictionary Iteration

## `.keys()`

````python
siswa = {"nama": "Andi", "umur": 21, "kota": "Jakarta"}

for k in siswa.keys():
    print(k)
````

## `.values()`

````python
siswa = {"nama": "Andi", "umur": 21, "kota": "Jakarta"}

for v in siswa.values():
    print(v)
````

## `.items()`

* Memberikan key dan value sekaligus

````python
for k, v in siswa.items():
    print(f"Key {k} - Values {v}")
````

Kombinasi dengan list dictionary

````python
data_karyawan = [
    {"nama": "Andi", "departemen": "IT"},
    {"nama": "Budi", "departemen": "Finance"},
]

for karyawan in data_karyawan:
    for field, isi in karyawan.items():
        print(f"{field} {isi}")
    print("--------------")
````

# 5. While Loop

## Condition-based Iteration

* While mengulang selama kondisinya True
* Kondisi pada while sama persis konsepnya dengan kondisi pada if (materi 07): hasil perbandingan, atau nilai yang dicek truthy/falsy-nya

````python
counter = 0

while counter < 5:
    print(counter)
    counter = counter + 1 # WAJIB update, supaya kondisi akhirnya jadi False
````

* `counter = counter + 1` → `counter += 1`
* `+=` disebut **augmented assignment**.
* Keduanya menghasilkan **hasil yang sama**.
* Bentuk lain: `-=`, `*=`, `/=`, dll.
* `+=` lebih ringkas dan **sering digunakan dalam kode**.


## Counter

* Pola "counter" (variabel penghitung yang naik/turun tiap putaran) adalah cara paling umum mengontrol kapan while berhenti
* Wajib diinisialisasi sebelum loop, dan diupdate di dalam loop

## Infinite loop

* Ini bug klasik: kalau kondisi while tidak pernah menjadi False, loop akan berjalan selamanya (program macet, harus dihentikan paksa)
* Penyebab paling umum: lupa update variabel yang dipakai dalam kondisi

````python
# CONTOH BUG -> JANGAN dijalankan seperti ini, akan macet selamanya
# counter = 0
# while counter < 5:
#     print(counter)
#     # lupa counter += 1 -> counter selamanya 0, kondisi selamanya True

# VERSI YANG BENAR -> selalu pastikan ada langkah yang membawa kondisi ke False
counter = 0
while counter < 5:
    print(counter)
    counter += 1   # baris ini yang mencegah infinite loop
````

> Kalau kode kamu "hang"/tidak berhenti-berhenti saat dijalankan, kemungkinan besar ini penyebabnya. Di Jupyter Notebook, kamu bisa tekan tombol "Stop" atau restart kernel untuk menghentikan paksa.

## Kapan menggunakan while

* Pakai for kalau kamu tahu persis apa yang mau diulang (koleksi data yang sudah ada, atau jumlah pengulangan yang sudah pasti)
* Pakai while kalau jumlah pengulangan tidak diketahui di awal — tergantung suatu kondisi yang baru terpenuhi belakangan

````python
# Contoh use-case while yang wajar: mengulang sampai syarat tertentu terpenuhi
saldo = 100000
biaya_admin = 15000

bulan = 0
while saldo > 0:
    saldo = saldo - biaya_admin
    bulan += 1

print(f"Saldo habis setelah {bulan} bulan")
# tidak bisa diketahui di awal berapa kali loop akan berjalan -> cocok pakai while
````

# 6. Loop Control

## break

* Menghentikan loop secara total, langsung keluar — sisa iterasi yang belum jalan tidak akan dieksekusi
* Paling umum dipakai untuk kasus pencarian: begitu ketemu, tidak perlu lanjut mengecek sisanya

````python
nama_siswa = ["Andi", "Budi", "Citra", "Dedi"]

for nama in nama_siswa:
    if nama == "Citra":
        print("Ditemukan!")
        break              # langsung berhenti, tidak perlu cek "Dedi" lagi
    print(f"Mengecek {nama}...")

# Output:
# Mengecek Andi...
# Mengecek Budi...
# Ditemukan!
````

## Continue

Melewati sisa kode di iterasi saat ini saja, lalu lanjut ke elemen berikutnya — loop tidak berhenti total, cuma "lompat" satu putaran

````python
angka = [1, 2, 3, 4, 5, 6, 7, 8]

for n in angka:
    if n % 2 != 0:      # kalau ganjil (modulus dari materi 02)
        continue          # lewati, jangan print, tapi lanjut ke elemen berikutnya
    print(n)
# 2
# 4
# 6
# 8
````

> Ringkasan perbedaan: break = "berhenti semuanya, keluar dari loop". continue = "lewati yang ini saja, lanjut ke berikutnya".

# 7. Loop Else

## Konsep else pada loop

* Blok else setelah for/while akan dieksekusi hanya jika loop selesai secara normal — TANPA pernah terkena break
* Kalau loop berhenti karena break, blok else dilewati sepenuhnya

### Use case sederhana

Paling umum dipakai untuk kasus: "cari sesuatu, dan kalau tidak ketemu sampai akhir, lakukan sesuatu"

````python
nama_siswa = ["Andi", "Budi", "Citra"]
cari = "Eka"

for nama in nama_siswa:
    if nama == cari:
        print(f"{cari} ditemukan!")
        break
else:
    # blok ini HANYA jalan kalau loop selesai TANPA break
    print(f"{cari} tidak ditemukan dalam daftar")

# Output: Eka tidak ditemukan dalam daftar
````

Bandingkan kalau nama-nya ditemukan:

````python
cari2 = "Budi"

for nama in nama_siswa:
    if nama == cari2:
        print(f"{cari2} ditemukan!")
        break
else:
    print(f"{cari2} tidak ditemukan dalam daftar")

# Output: Budi ditemukan!
# -> blok else TIDAK dieksekusi, karena loop berhenti lewat break
````

## Awareness terhadap syntax

* Indentasi else sejajar dengan for/while, bukan menjorok ke dalam seperti if-else biasa — ini sering bikin bingung karena namanya sama (else) tapi cara kerjanya beda total dari else pada materi 07
* Kalau loop tidak punya break sama sekali, blok else-nya akan selalu jalan (karena "selesai tanpa break" otomatis benar)
* Fitur ini opsional — kebanyakan kasus pencarian sebenarnya bisa juga ditulis pakai variabel boolean penanda (ditemukan = False, ubah jadi True kalau ketemu), tanpa else pada loop. Tapi mengenali sintaks ini penting supaya tidak bingung kalau menemukan di kode orang lain.
