# Bagian A — Tuple

## 1. Konsep Tuple

* Tupple: struktur data yang menyimpan kumpulan nilai, mirip list, tetapi mempunyai sifat dasar berbeda: tidak bisa diubah setelah dibuat
* Ditulis menggunakan tanda kurung biasa ( ), tiap elemen dipisahkan koma
* Sering dipakai untuk data yang secara konsep memang "sudah final" — misalnya satu titik koordinat (x, y) yang tidak masuk akal kalau tiba-tiba salah satu angkanya berubah sendiri

````python
# Tuple dengan beberapa elemen
koordinat = (10, 20)
data_siswa = ("Andi", 20, "Jakarta")

# Tuple dengan satu elemen -> WAJIB pakai koma di belakang!
tuple_satu = (5,)      # ini tuple
bukan_tuple = (5)      # ini cuma int biasa dalam kurung, BUKAN tuple

print(type(tuple_satu))    # <class 'tuple'>
print(type(bukan_tuple))   # <class 'int'>

# Tuple juga bisa dibuat tanpa kurung sama sekali (parentheses opsional)
tuple_tanpa_kurung = 1, 2, 3
print(type(tuple_tanpa_kurung))   # <class 'tuple'>

# Tuple kosong
tuple_kosong = ()
````

> Jebakan paling umum: lupa koma untuk tuple satu elemen. (5) itu cuma angka 5 dalam kurung biasa, bukan tuple — ini sering bikin bug yang sulit ditemukan karena tidak ada error yang jelas.

Ordered collection
* Sama seperti list, tuple mempertahankan urutan sesuai saat dibuat
* Karena berurutan, tiap elemen bisa diakses lewat index

Immutable collection
* Ini perbedaan paling mendasar dari list: sekali dibuat, isi tuple tidak bisa diubah — tidak bisa tambah, hapus, atau ganti elemen
* Kalau butuh "versi lain" dari tuple, kamu harus membuat tuple baru, bukan mengubah yang lama

```python
Ini akan ERROR:

data[0] = "Budi"  

 -> TypeError: 'tuple' object does not support item assignment

````python
data = ("Andi", 20, "Jakarta")

# Cara yang benar kalau butuh "versi baru": buat tuple baru
data_baru = ("Budi",) + data[1:]
print(data_baru)   # ('Budi', 20, 'Jakarta')
````

## 2. Operasi Tuple

Karena tuple itu ordered (berurutan), sebagian besar cara mengakses datanya identik dengan list — bedanya cuma di soal bisa-tidaknya diubah.

Indexing

````python
data = ("Andi", 20, "Jakarta", True)
#          0      1       2        3

print(data[0])    # 'Andi'
print(data[-1])   # True -> negative indexing juga berlaku
````

Slicing

````python
angka = (10, 20, 30, 40, 50)

print(angka[1:4])    # (20, 30, 40) -> hasil slicing tuple, tetap berupa tuple
print(angka[::-1])   # (50, 40, 30, 20, 10) -> dibalik
````

len()

````python
data = ("Andi", 20, "Jakarta")
print(len(data))   # 3
````

Membership in

````python
data = ("Andi", 20, "Jakarta")

print("Andi" in data)       # True
print("Budi" in data)       # False
print("Budi" not in data)   # True
````

> Poin penting: tuple tidak punya method seperti .append(), .remove(), .sort(), dll — karena semua method itu fungsinya mengubah isi, sedangkan tuple memang dirancang untuk tidak bisa diubah. Method yang tersedia di tuple cuma .count() dan .index() (sama fungsinya seperti di list, hanya untuk "bertanya", bukan "mengubah").

## 3. Tuple Unpacking

Basic unpacking

* Salah satu fitur paling elegan dari tuple: kamu bisa "membongkar" isi tuple langsung ke beberapa variabel sekaligus
* Jumlah variabel di kiri harus sama persis dengan jumlah elemen tuple, kalau tidak akan error

````python
data_siswa = ("Andi", 20, "Jakarta")

# Unpacking -> masing-masing elemen otomatis masuk ke variabel sesuai urutan
nama, umur, kota = data_siswa

print(nama)   # 'Andi'
print(umur)   # 20
print(kota)   # 'Jakarta'
````

Multiple assignment menggunakan tuple
* Konsep unpacking ini juga yang memungkinkan trik "tukar nilai variabel" tanpa variabel bantuan, dan return value ganda dari fungsi

````python
# Trik swap nilai tanpa variabel sementara -> ini sebenarnya unpacking tuple di balik layar
a = 5
b = 10
a, b = b, a
print(a, b)   # 10 5

# Fungsi yang me-return beberapa nilai sekaligus, sebenarnya me-return satu tuple
def bagi_dan_sisa(a, b):
    hasil_bagi = a // b
    sisa = a % b
    return hasil_bagi, sisa   # ini otomatis jadi tuple (hasil_bagi, sisa)

hasil, sisa = bagi_dan_sisa(17, 5)
print(hasil, sisa)   # 3 2

# Unpacking sebagian dengan * (menampung "sisanya" ke satu variabel list)
angka = (1, 2, 3, 4, 5)
pertama, *tengah, terakhir = angka
print(pertama)    # 1
print(tengah)     # [2, 3, 4] -> ditampung sebagai list
print(terakhir)   # 5
````

## 4. Tuple vs List

Mutability
* List → mutable, isinya bisa diubah kapan saja
* Tuple → immutable, sekali dibuat isinya tetap selamanya

Use-case
* List cocok untuk kumpulan data yang memang diharapkan berubah — misal daftar hasil query yang terus bertambah, list nilai siswa yang bisa diupdate
* Tuple cocok untuk data yang secara konsep "satu paket yang final" — misal koordinat (latitude, longitude), RGB warna (255, 0, 0), atau baris data yang tidak boleh sengaja termodifikasi

Kapan menggunakan tuple?
* Saat data tidak seharusnya diubah setelah dibuat (mencegah bug tidak sengaja mengubah data penting)
* Saat butuh key untuk dictionary — list tidak bisa dipakai sebagai key dictionary karena mutable, tapi tuple bisa
* Saat fungsi perlu return beberapa nilai sekaligus
* Tuple sedikit lebih cepat dan hemat memori dibanding list untuk data dalam jumlah besar yang tidak berubah (bukan concern utama saat belajar, tapi baik diketahui)

````python
# Contoh: tuple sebagai key dictionary (list tidak bisa!)
jarak_kota = {
    ("Jakarta", "Bandung"): 150,
    ("Jakarta", "Surabaya"): 800
}
print(jarak_kota[("Jakarta", "Bandung")])   # 150

# Ini akan ERROR kalau pakai list sebagai key:
# jarak_kota[["Jakarta", "Bandung"]] = 150   -> TypeError: unhashable type: 'list'
````

# Bagian B — Set

## 5. Konsep Set

Unique elements (elemen unik)
* Set adalah struktur data yang hanya menyimpan nilai-nilai unik — otomatis membuang duplikat
* Ini yang bikin set sangat berguna untuk urusan deduplikasi data

Unordered nature (secara konseptual)
* Set tidak punya urutan yang bisa diandalkan — beda dengan list/tuple yang urutannya tetap sesuai input
* Konsekuensinya: set tidak bisa diakses lewat index (set[0] akan error)
* Karena tidak berurutan, saat kamu print() sebuah set, urutan tampilannya bisa berbeda dari urutan saat dimasukkan

````python
# Set dari beberapa nilai -> pakai kurung kurawal { }
buah = {"apel", "jeruk", "mangga"}

# Duplikat otomatis dibuang saat set dibuat
angka = {1, 2, 2, 3, 3, 3, 4}
print(angka)   # {1, 2, 3, 4} -> duplikat hilang otomatis

# Set kosong -> HARUS pakai set(), bukan {}
set_kosong = set()      # ini set kosong yang benar
bukan_set = {}          # ini sebenarnya dictionary kosong, BUKAN set!

print(type(set_kosong))   # <class 'set'>
print(type(bukan_set))    # <class 'dict'>

# Membuat set dari list (cara umum untuk deduplikasi)
list_dengan_duplikat = [1, 2, 2, 3, 3, 3]
set_unik = set(list_dengan_duplikat)
print(set_unik)   # {1, 2, 3}
````

> Jebakan umum: {} itu dictionary kosong, bukan set kosong! Kalau butuh set kosong, wajib pakai set().

## 6. Manipulasi Set

Menambah elemen

````python
buah = {"apel", "jeruk"}

# .add() -> tambah satu elemen
buah.add("mangga")
print(buah)   # {'apel', 'jeruk', 'mangga'} (urutan tampil bisa berbeda-beda)

# Menambahkan elemen yang sudah ada -> tidak berpengaruh, tidak error
buah.add("apel")
print(buah)   # tetap {'apel', 'jeruk', 'mangga'}, tidak ada duplikat baru

# .update() -> tambah banyak elemen sekaligus (mirip .extend() pada list)
buah.update(["pisang", "anggur"])
print(buah)   # {'apel', 'jeruk', 'mangga', 'pisang', 'anggur'}
````

Menghapus elemen

````python
buah = {"apel", "jeruk", "mangga"}

# .remove() -> hapus elemen tertentu, ERROR kalau elemen tidak ada
buah.remove("jeruk")
print(buah)   # {'apel', 'mangga'}

# .discard() -> sama seperti remove(), tapi TIDAK error kalau elemen tidak ditemukan
buah.discard("melon")   # tidak ada 'melon', tapi tidak error
print(buah)               # {'apel', 'mangga'} -> tetap aman

# .pop() -> hapus dan kembalikan satu elemen secara acak (karena set tidak berurutan)
elemen_terhapus = buah.pop()
print(elemen_terhapus)   # bisa 'apel' atau 'mangga', tidak bisa dipastikan

# .clear() -> kosongkan seluruh set
buah.clear()
print(buah)   # set()
````

> .remove() vs .discard(): kalau kamu tidak yakin elemennya ada atau tidak, pakai .discard() supaya tidak error. Kalau kamu yakin elemennya pasti ada dan mau tahu kalau ternyata salah (bug), pakai .remove().

Membership in

````python
buah = {"apel", "jeruk", "mangga"}

print("apel" in buah)      # True
print("melon" in buah)     # False
````

> Kenapa penting: mengecek keberadaan elemen (in) di dalam set jauh lebih cepat dibanding di dalam list, terutama untuk data dalam jumlah besar. Ini salah satu alasan set sering dipakai saat butuh pengecekan keanggotaan berulang kali

## 7. Set Operations

Union (gabungan) — | atau .union()
* Menggabungkan semua elemen dari kedua set, duplikat otomatis hilang

````python
kelas_a = {"Andi", "Budi", "Citra"}
kelas_b = {"Citra", "Dedi", "Eka"}

gabungan = kelas_a | kelas_b
# atau: gabungan = kelas_a.union(kelas_b)

print(gabungan)   # {'Andi', 'Budi', 'Citra', 'Dedi', 'Eka'} -> semua siswa, tanpa duplikat
````

Intersection (irisan) — & atau .intersection()
* Hanya elemen yang muncul di kedua set

````python
irisan = kelas_a & kelas_b
# atau: irisan = kelas_a.intersection(kelas_b)

print(irisan)   # {'Citra'} -> siswa yang ada di kedua kelas
````

Difference (selisih) — - atau .difference()
* Elemen yang ada di set pertama, tapi tidak ada di set kedua
* Perhatikan: urutan set berpengaruh ke hasil!

````python
selisih_a = kelas_a - kelas_b
print(selisih_a)   # {'Andi', 'Budi'} -> hanya ada di kelas_a, tidak di kelas_b

selisih_b = kelas_b - kelas_a
print(selisih_b)   # {'Dedi', 'Eka'} -> hanya ada di kelas_b, tidak di kelas_a
````

Symmetric difference (selisih simetris) — ^ atau .symmetric_difference()
*  Elemen yang hanya ada di salah satu set (bukan di keduanya) — kebalikan dari intersection

````python
sym_diff = kelas_a ^ kelas_b
# atau: sym_diff = kelas_a.symmetric_difference(kelas_b)

print(sym_diff)   # {'Andi', 'Budi', 'Dedi', 'Eka'} -> semua KECUALI yang di kedua kelas ('Citra')
````

Ringkasan cara mengingat:

* Union = gabungkan semua
* Intersection = yang sama-sama punya
* Difference = yang cuma dimiliki satu pihak (arahnya penting: A-B ≠ B-A)
* Symmetric difference = yang unik masing-masing (gabungan minus irisan)

## 8. Use Case

Menghilangkan duplicate

* Ini adalah kegunaan paling umum dari set dalam praktik data science
* Cara tercepat menghapus duplikat dari sebuah list: ubah jadi set, lalu (kalau perlu) ubah balik ke list

````python
data_id_pelanggan = [101, 102, 103, 101, 104, 102, 105]

id_unik = list(set(data_id_pelanggan))
print(id_unik)   # [Terurut atau tidak, tergantung Python, isinya: 101,102,103,104,105]

# Hitung berapa banyak ID unik
print(f"Jumlah pelanggan unik: {len(set(data_id_pelanggan))}")   # 5
````

> Catatan: mengubah list ke set lalu balik ke list tidak menjaga urutan asli. Kalau urutan penting, ini bukan cara yang tepat (nanti akan ada teknik lain saat belajar Pandas, misal .drop_duplicates()).

Membandingkan collection
* Set sangat berguna untuk pertanyaan bisnis semacam "siapa saja yang ada di kedua data" atau "siapa yang cuma ada di salah satu"

````python
pelanggan_bulan_ini = {"Andi", "Budi", "Citra", "Dedi"}
pelanggan_bulan_lalu = {"Budi", "Citra", "Eka"}

# Pelanggan baru (ada bulan ini, tidak ada bulan lalu)
pelanggan_baru = pelanggan_bulan_ini - pelanggan_bulan_lalu
print(f"Pelanggan baru: {pelanggan_baru}")   # {'Andi', 'Dedi'}

# Pelanggan yang hilang (ada bulan lalu, tidak ada bulan ini)
pelanggan_hilang = pelanggan_bulan_lalu - pelanggan_bulan_ini
print(f"Pelanggan hilang: {pelanggan_hilang}")   # {'Eka'}

# Pelanggan setia (ada di kedua bulan)
pelanggan_setia = pelanggan_bulan_ini & pelanggan_bulan_lalu
print(f"Pelanggan setia: {pelanggan_setia}")   # {'Budi', 'Citra'}
````
