# 1. Konsep Function

## Apa itu function?

* Function : blok kode yang diberi nama, bisa dipanggil untuk dijalankan kapan saja, sebanyak apa pun, tanpa perlu menulis ulang kodenya
* print(), len(), type(), sorted() semuanya adalah function bawaan Python. Bedanya, sekarang akan belajar membuat function sendiri

## Mengapa function digunakan?

* Menghindari duplikasi kode — kalau logika yang sama dipakai di banyak tempat, cukup tulis sekali dalam function, lalu panggil berkali-kali
* Membuat kode lebih mudah dites dan diperbaiki — kalau ada bug, kamu cukup perbaiki di satu tempat (dalam function), bukan mencari-cari di banyak tempat kode yang sama ditulis ulang
* Membuat kode lebih mudah dibaca — nama function yang deskriptif (hitung_diskon(), bukan sekumpulan baris tanpa nama) langsung menjelaskan maksud kode tanpa perlu membaca isinya detail

## Reusability

Sekali function dibuat, kamu bisa memanggilnya berkali-kali dengan input berbeda-beda, tanpa menulis ulang logikanya

```python
# TANPA function -> logika yang sama ditulis berulang, rawan typo/inkonsisten
harga1 = 100000
diskon1 = 100000 - (100000 * 10 / 100)
print(diskon1)

harga2 = 50000
diskon2 = 50000 - (50000 * 10 / 100)
print(diskon2)

# DENGAN function -> logika ditulis sekali, dipakai berkali-kali
def hitung_harga_diskon(harga, persen_diskon):
    return harga - (harga * persen_diskon / 100)

print(hitung_harga_diskon(100000, 10))   # 90000.0
print(hitung_harga_diskon(50000, 10))    # 45000.0
```

keyword `def` (define) menandai awal pembuatan fanction baru

# 3. Parameter & Argument

## Parameter

* nama varibel yang dituliskan dalam kurung saat mendefinisikan function - seperti slot kosong yang akan diisi

## Argument

* Nilai sebenarnya yang akan dimasukan ke slot saat memanggil function

```python
# 'nama' dan 'umur' di sini adalah PARAMETER
def perkenalan(nama, umur):
    print(f"Nama saya {nama}, umur {umur} tahun")

# "Andi" dan 25 di sini adalah ARGUMENT
perkenalan("Andi", 25)   # Nama saya Andi, umur 25 tahun
```

## Positional argument

* Argument yang dikirim berdasarkan urutan posisinya, mengikuti urutan parameter di definisi
* Urutan sangat penting — tertukar urutan, hasilnya ikut salah (tanpa error, ini yang bahaya)

```python
def perkenalan(nama, umur):
    print(f"Nama saya {nama}, umur {umur} tahun")

perkenalan("Andi", 25)   # benar: nama="Andi", umur=25
perkenalan(25, "Andi")   # SALAH URUTAN, tapi TIDAK error!
# Output: "Nama saya 25, umur Andi tahun" -> logikanya kacau, tapi Python tetap jalan
```

## Keyword argument

Argument yang dikirim dengan menyebutkan nama parameternya secara eksplisit — urutan jadi tidak masalah, karena Python tahu persis argument mana untuk parameter mana

```python
# Keyword argument -> urutan boleh dibalik, hasilnya tetap benar
perkenalan(umur=25, nama="Andi")   # Nama saya Andi, umur 25 tahun -> tetap benar!

# Bisa juga dicampur: positional dulu, baru keyword
perkenalan("Andi", umur=25)   # valid

# Tapi TIDAK BOLEH keyword dulu baru positional
# perkenalan(nama="Andi", 25)   -> SyntaxError!
```

> Kapan pakai keyword argument? Sangat disarankan dipakai kalau function punya banyak parameter (3 atau lebih) — supaya orang yang membaca kode (termasuk kamu sendiri nanti) langsung tahu maksud tiap argument tanpa perlu buka definisi function-nya.

# 4. Default Parameter

## Default value

Parameter bisa diberi nilai bawaan — kalau saat memanggil function argument-nya tidak diisi, nilai default inilah yang dipakai

```python
def perkenalan(nama, kota="Surabaya"):
    print(f"Nama saya {nama} dan kota {kota}")

perkenalan("Andi", "Jakarta")
perkenalan("Bagus")
```

## Penggunaan default parameter

* Aturan penting: parameter yang punya default harus diletakkan setelah parameter yang tidak punya default — kalau dibalik, akan error

```python
# BENAR -> parameter tanpa default duluan, baru yang punya default
def hitung_diskon(harga, persen=10):
    return harga - (harga * persen / 100)

print(hitung_diskon(100000))       # 90000.0 -> persen pakai default 10
print(hitung_diskon(100000, 20))   # 80000.0 -> persen di-override jadi 20

# SALAH -> ini akan error:
# def hitung_diskon(persen=10, harga):
#     ...
# -> SyntaxError: non-default argument follows default argument
```

# 5. Return

## `return`

* keyboard untuk mengirim nilai keluar dari function supaya bisa dipakai/disimpan di luar function tersebut
* Begitu return dieksekusi, function langsung berhenti — kode setelah return di dalam function yang sama tidak akan dijalankan

```python
def tambah(a, b):
    return a + b
    print("Ini tidak akan pernah tercetak")   # kode setelah return diabaikan

hasil = tambah(5, 3)
print(hasil)   # 8
```

## Return value

Nilai yang di-return bisa disimpan ke variabel untuk dipakai lagi, beda dengan sekadar print() yang cuma menampilkan ke layar (perbedaan detailnya di bagian selanjutnya)

```python
def kuadrat(angka):
    return angka ** 2   # ** adalah operator pangkat dari materi 02

hasil_kuadrat = kuadrat(5)
print(hasil_kuadrat + 10)   # 35 -> hasil return bisa dipakai untuk operasi lain
```

## Multiple return values

* Function bisa me-return lebih dari satu nilai sekaligus, dipisah koma — di balik layar, Python sebenarnya membungkusnya jadi satu tuple (ingat konsep tuple dan unpacking dari materi 04!)

```python
def bagi_dan_sisa(a, b):
    hasil_bagi = a // b   # floor division dari materi 02
    sisa = a % b            # modulus dari materi 02
    return hasil_bagi, sisa   # ini otomatis jadi tuple: (hasil_bagi, sisa)

# Unpacking langsung ke dua variabel -> persis konsep materi 04
hasil, sisa = bagi_dan_sisa(17, 5)
print(hasil)   # 3
print(sisa)    # 2

# Kalau ditampung satu variabel saja, akan terlihat bentuk tuple aslinya
gabungan = bagi_dan_sisa(17, 5)
print(gabungan)         # (3, 2)
print(type(gabungan))   # <class 'tuple'>
```

## return vs print()

| Aspek                      | `print()`                       | `return`                                        |
| -------------------------- | ------------------------------- | ----------------------------------------------- |
| Fungsi                     | Menampilkan teks/nilai ke layar | Mengirim nilai keluar dari function             |
| Bisa disimpan ke variabel? | ❌ Tidak — hasilnya `None`       | ✅ Ya — nilainya bisa ditangkap dan dipakai lagi |
| Menghentikan function?     | ❌ Tidak                         | ✅ Ya, function langsung berhenti                |


```python
def tambah_print(a, b):
    print(a + b)      # cuma MENAMPILKAN, tidak mengirim nilai keluar

hasil1 = tambah_print(5, 3)    # tercetak "8" ke layar SAAT function dipanggil
print(hasil1)                    # None! -> karena tambah_print tidak return apa pun
print(hasil1 + 10)  # ERROR! -> TypeError, None tidak bisa dijumlah
```

```python
def tambah_return(a, b):
    return a + b       # mengirim nilai keluar, bisa dipakai lagi             

hasil2 = tambah_return(5, 3)   # TIDAK ada yang tercetak saat function dipanggil
print(hasil2)                    # 8 -> nilainya berhasil disimpan
print(hasil2 + 10)               # 18 -> bisa dipakai untuk operasi lain
```

> Prinsip mengingat: print() itu untuk manusia (menampilkan sesuatu supaya bisa dibaca). return itu untuk program (mengirim nilai supaya bisa diproses lebih lanjut oleh kode lain). Function yang dirancang untuk menghasilkan sesuatu yang akan dipakai lagi (menghitung, memvalidasi, mentransformasi) harus pakai return, bukan cuma print().

# 6. Basic Function Design

## Input → Process → Output

* Cara paling sederhana membayangkan function yang baik: dia menerima sesuatu (input via parameter), melakukan sesuatu (proses di dalam body), lalu menghasilkan sesuatu (output via return)
* Pola pikir ini membantu banget saat kamu bingung mau mulai menulis function dari mana

```python
def hitung_rata_rata(daftar_angka):
    total = 0
    for angka in daftar_angka:
        total = total + angka
    rata_rata = total / len(daftar_angka)
    return rata_rata

nilai = [80, 90, 75, 85, 95]
print(hitung_rata_rata(nilai))
```

## Function sederhana

* Function tidak harus rumit untuk berguna — bahkan function satu baris yang memberi nama pada sebuah operasi sudah meningkatkan keterbacaan kode

```python
def is_genap(angka):
    return angka % 2 == 0   # langsung return hasil perbandingan (sudah bertipe boolean)

print(is_genap(4))   # True
print(is_genap(7))   # False
```

## Single responsibility (pengantar)

* Prinsip ini bilang: satu function sebaiknya mengerjakan satu tugas saja, dan mengerjakannya dengan baik
* Kalau function mulai melakukan banyak hal berbeda sekaligus (validasi dan hitung dan cetak laporan dalam satu function), function itu jadi susah dites dan susah dipakai ulang untuk kebutuhan lain

```python
# Kurang baik satu function melakukan hal terlalu banyak

def proses_nilai_siswa(nilai):
    if nilai < 0 or nilai >100:
        print("Nilai tidak valid")
        return None
    if nilai > 60:
        status = "Lulus"
    else:
        status = "Tidak Lulus"
    print(f"Status siswa: {status}")
    return status

proses_nilai_siswa(6)
```

```python
# KURANG BAIK -> satu function melakukan terlalu banyak hal sekaligus
def proses_nilai_siswa(nilai):
    if nilai < 0 or nilai > 100:
        print("Nilai tidak valid")
        return None
    if nilai >= 60:
        status = "Lulus"
    else:
        status = "Tidak Lulus"
    print(f"Status: {status}")   # mencetak DAN memvalidasi DAN menentukan status
    return status
```

```python
# LEBIH BAIK -> dipecah, masing-masing punya satu tanggung jawab jelas
def is_nilai_valid(nilai):
    return 0 <= nilai <= 100   # bisa dibaca seperti notasi matematika biasa

def tentukan_status(nilai):
    return "Lulus" if nilai >= 60 else "Tidak Lulus"   # ternary dari materi 07

# Dipakai bersama, tapi masing-masing bisa dites/dipakai terpisah juga
nilai_siswa = 75
if is_nilai_valid(nilai_siswa):
    print(tentukan_status(nilai_siswa))   # Lulus
else:
    print("Nilai tidak valid")
```

> Kenapa versi kedua lebih baik: is_nilai_valid() dan tentukan_status() sekarang bisa dipakai ulang secara terpisah di bagian lain program, dan lebih gampang dites satu-satu. Ini pengantar konsep yang akan terus relevan sampai ke level pipeline data science — nanti di materi 1.7 (Data Cleaning Framework) kamu akan menulis banyak function kecil seperti handle_missing(), remove_duplicates() yang masing-masing fokus satu tugas, persis prinsip ini.

1. Function menghitung rata-rata

    Buat function hitung_rata_rata(daftar_angka) yang me-return rata-rata dari list angka yang diberikan (gunakan for loop, bukan sum()). Uji dengan minimal 2 list angka berbeda, simpan hasilnya ke variabel, lalu gunakan hasil itu untuk perhitungan lain (misal dibulatkan atau dibandingkan dengan angka lain) — untuk membuktikan function ini benar return, bukan cuma print().

```python
def hitung_rata_rata(daftar_angka):
    total = 0
    for angka in daftar_angka:
        total = total + angka
    rata_rata = total / len(daftar_angka)
    return rata_rata

angka1 = [12, 32, 12, 31, 26, 32]
angka2 = [32, 12, 34, 15, 12, 42]

avg1 = hitung_rata_rata(angka1)
avg2 = hitung_rata_rata(angka2)
print(avg1)
print(avg2)
    
    
if avg1 > avg2:
    print("Ok")
else:
    print("Ok")
```

2. Function filtering
    
    Buat function filter_lulus(daftar_nilai, batas_lulus=60) yang menerima list nilai, lalu me-return list baru berisi nilai-nilai yang >= batas_lulus (gunakan for + if, parameter batas_lulus punya default value 60). Panggil function ini dua kali: sekali tanpa mengisi batas_lulus (pakai default), sekali lagi dengan mengisi batas_lulus secara eksplisit sebagai keyword argument.

```python
def filter_lulus(daftar_nilai, batas_lulus=60):
    nilai_lulus = []
    for nilai in daftar_nilai:
        if nilai > batas_lulus:
            nilai_lulus.append(nilai)
            
    return nilai_lulus

nilai1 = [55, 65, 50, 80, 90]

hasil_nilai = filter_lulus(nilai1)
hasil_nilai_70 = filter_lulus(nilai1, 70)
print(hasil_nilai)
print(hasil_nilai_70)
```

3. Function validasi
    
    Buat function validasi_umur(umur) yang me-return True kalau umur masuk akal (antara 0 sampai 120), dan False kalau tidak. Lalu buat function kedua kategori_umur(umur) yang me-return kategori umur ("Anak-anak"/"Remaja"/"Dewasa"/"Lansia", pakai logika dari materi 07). Gabungkan keduanya: hanya panggil kategori_umur() kalau validasi_umur() bernilai True — ini contoh penerapan single responsibility dari bagian 6.

```python
def validasi_umur(umur):
    """Memvalidasi apakah umur berada dalam rentang 0-120."""
    return 0 <= umur <= 120


def kategori_umur(umur):
    """Menentukan kategori berdasarkan umur."""
    if umur <= 12:
        return "Anak-anak"
    elif umur <= 17:
        return "Remaja"
    elif umur <= 59:
        return "Dewasa"
    else:
        return "Lansia"


umur = 140

if validasi_umur(umur):
    print(kategori_umur(umur))
else:
    print("Umur tidak valid")
```

4. Function transformasi sederhana
    
    Buat function format_rupiah(angka) yang menerima angka, lalu me-return string berformat rupiah, misal input 150000 menghasilkan "Rp150,000" (gunakan f-string dengan format pemisah ribuan dari materi 02). Lalu buat function kedua terapkan_diskon(harga, persen_diskon=0) yang me-return dua nilai sekaligus (multiple return): harga setelah diskon, dan besar potongan dalam rupiah. Tangkap kedua return value itu dengan unpacking ke dua variabel terpisah, lalu tampilkan keduanya memakai format_rupiah() yang sudah kamu buat.

```python
def format_rupiah(angka):
    return f"Rp{angka:,.0f}"


def terapkan_diskon(harga, persen_diskon=0):
    potongan = harga * persen_diskon / 100
    harga_akhir = harga - potongan

    return harga_akhir, potongan


# Contoh penggunaan
harga = 150000
persen_diskon = 20

harga_setelah_diskon, besar_potongan = terapkan_diskon(
    harga, persen_diskon
)

print("Harga setelah diskon:", format_rupiah(harga_setelah_diskon))
print("Besar potongan:", format_rupiah(besar_potongan))
```
