# Exercise 2 Simple Calculation

```python
price = 25000
quantity = 4
discount = 0.10

subtotal = price * quantity
discount_amount = subtotal * discount
total = subtotal - discount_amount

print(f"Subtotal: {subtotal}")
print(f"Discount Amount: {discount_amount}")
print(f"Total: {total}")
```
