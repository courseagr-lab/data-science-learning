## 2.List

### 2.1 Konsep dasar & Cara kerja internal

* List = Struktur data berupa kumpulan elemen yang dinamis
* Elemen disimpan dalam blok memory yang dapat bertambah
* Saat kapasitas penuh dan elemen baru ditambahkan:
    * Python mengalokasikan blok memory yang lebih besar
    * Elemen lama disimpan ke likasi baru
    * Elemen baru kemudian ditambahkan
* Tujuannya agar list dapat bertumbuh secara fleksibel tanpa menentukan ukuran sejak awal

Implikasinya: append() ke list sangat cepat secara rata-rata (O(1) amortized), tapi insert() di tengah lambat (O(n)) karena semua elemen setelahnya harus digeser.

````python
# Membuat list
empty = []
empty2 = list()

numbers = [1, 2, 3, 4, 5]
mixed = [1, 'python', 3.14, True, None] # Bisa berisi tipe data yang berbeda
nested = [[1,2], [3, 4], [5, 6]] # list of list

print(mixed)
````

````python
# List dari interable lain
from_range = list(range(10)) #
from_string = list("Python") # ['P', 'y', 't', 'h', 'o', 'n']
from_tuple = list((1, 2, 3)) # [1, 2, 3]

print(from_range)
print(from_string)
print(from_tuple)
````

### 2.2 Indexing & Sliccing

List menggunakan mekanisme indexing dan slicing yang identik dengan string, karena keduanya adalah sequence

````python
data = [10, 20, 30, 40, 50, 60, 70]

print("Indexing")
print(data[0])
print(data[-1])
print(data[-2])

print("\nSlicing")
print(data[1:4]) # [20, 30, 40]
print(data[:3])
print(data[4:])
print(data[::2])
print(data[::-1])
````

List bisa dimodifikasi melalui indexinf & slicing

````python
data = [10, 20, 30, 40, 50]

# Ubah satu elemen
data[0] = 99
print(data)
````

````python
# Ubah slice sekaligus
data[1:3] = [200, 300]
print(data)
````

````python

# Hapus dengan slice
data[1:3] = []
print(data)
````

````python
data[1:1] = [11, 12, 13] # Menyisipkan elemen baru
print(data)
````
