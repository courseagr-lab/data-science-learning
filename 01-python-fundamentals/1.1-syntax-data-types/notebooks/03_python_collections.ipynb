## 2.1 Konsep dasar & Cara kerja internal

* List = Struktur data berupa kumpulan elemen yang dinamis
* Elemen disimpan dalam blok memory yang dapat bertambah
* Saat kapasitas penuh dan elemen baru ditambahkan:
    * Python mengalokasikan blok memory yang lebih besar
    * Elemen lama disimpan ke likasi baru
    * Elemen baru kemudian ditambahkan
* Tujuannya agar list dapat bertumbuh secara fleksibel tanpa menentukan ukuran sejak awal

Implikasinya: append() ke list sangat cepat secara rata-rata (O(1) amortized), tapi insert() di tengah lambat (O(n)) karena semua elemen setelahnya harus digeser.

## 2.4 Copy — Shallow vs Deep

````python
# Assignment - bukan copy, hanya alisa

original = [1, 2, 3]
alias = original

alias[0] = 99
print(original) # Ikut berubah
print(alias)
````

original dan alias menunjuk objek yang sama pada memory, mengubah satu berarti mengubah keduannya

````python
# Shallow copy - tiga cara yang setara

original = [1, 2, 3]

copy1 = original.copy()
copy2 = original[:]
copy3 = list(original)


copy1[0] = 99
print(original)  # Tidak berubah
print(copy1)
````

Shallow copy cukup untuk list satu dimensi. Masalah muncul dengan list bersarang

````python
# Masalah shallow copy dengan list bersarang

import copy
nested = [[1, 2], [3, 4], [5, 6]]
shallow = nested.copy()

shallow[0][0] = 99
print(nested) # Nested ikut berubah
print(shallow)
````

Shallow copy hanya menyalin referensi ke objek didalam list, bukan object itu sendiri. List dalam innner list masih berbagi memory

````python
# Deep copy - menyalin seluruh struktur secara rekrusif

nested = [[1, 2], [3, 4], [5, 6]]
deep = copy.deepcopy(nested)

deep[0][0] = 99
print(nested)  # Tidak ikut berubah
print(deep)
````

### Repetisi

````python
repeated = [0] * 5
print(repeated)
````

````python
# Repitisi dengan list bersarang

matrix = [[0] * 3] * 3
matrix[0][0] = 99
print(matrix)

# Semua baris adalah objek yang sama
````

````python
# Cara yang benar

matrix = [[0] * 3 for _ in range(3)]
matrix[0][0] = 99
print(matrix)
````

## 2.6 Membership dan Iterasi

````python
data = [10, 20, 30, 40, 50]

# Membership test
print(30 in data)
print(99 in data)
print(99 not in data)
````

````python
# Iterasi dasar
for item in data:
    print(item)
````

````python
# Iterasi dengan index

for i, item in enumerate(data):
    print(f"Index {i}: {item}")
````

````python
# Iterasi dengan dua list bersamaan

names = ['Andi', 'bagus', 'rahmat']
scores = [40, 34, 50]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# zip berhenti di list yang lebih pendek
# gunakan zip_longest dari itertools jika perlu
````
