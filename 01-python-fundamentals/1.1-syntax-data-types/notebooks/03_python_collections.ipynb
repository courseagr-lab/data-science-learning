# 2. List

## 2.1 Konsep dasar & Cara kerja internal

* List = Struktur data berupa kumpulan elemen yang dinamis
* Elemen disimpan dalam blok memory yang dapat bertambah
* Saat kapasitas penuh dan elemen baru ditambahkan:
    * Python mengalokasikan blok memory yang lebih besar
    * Elemen lama disimpan ke likasi baru
    * Elemen baru kemudian ditambahkan
* Tujuannya agar list dapat bertumbuh secara fleksibel tanpa menentukan ukuran sejak awal

Implikasinya: append() ke list sangat cepat secara rata-rata (O(1) amortized), tapi insert() di tengah lambat (O(n)) karena semua elemen setelahnya harus digeser.

## 2.4 Copy — Shallow vs Deep

````python
# Assignment - bukan copy, hanya alisa

original = [1, 2, 3]
alias = original

alias[0] = 99
print(original) # Ikut berubah
print(alias)
````

original dan alias menunjuk objek yang sama pada memory, mengubah satu berarti mengubah keduannya

````python
# Shallow copy - tiga cara yang setara

original = [1, 2, 3]

copy1 = original.copy()
copy2 = original[:]
copy3 = list(original)


copy1[0] = 99
print(original)  # Tidak berubah
print(copy1)
````

Shallow copy cukup untuk list satu dimensi. Masalah muncul dengan list bersarang

````python
# Masalah shallow copy dengan list bersarang

import copy
nested = [[1, 2], [3, 4], [5, 6]]
shallow = nested.copy()

shallow[0][0] = 99
print(nested) # Nested ikut berubah
print(shallow)
````

Shallow copy hanya menyalin referensi ke objek didalam list, bukan object itu sendiri. List dalam innner list masih berbagi memory

````python
# Deep copy - menyalin seluruh struktur secara rekrusif

nested = [[1, 2], [3, 4], [5, 6]]
deep = copy.deepcopy(nested)

deep[0][0] = 99
print(nested)  # Tidak ikut berubah
print(deep)
````

* List menyimpan koleksi item yang sejenis dan jumlahnya bisa berubah

````python
# List - koleksi item sejenis
scores = [12, 32, 12, 21]


# Tuple - satu entitas dengan beberapa atribut
employee = ("E001", "Andi Pratama", "Engineer", 7_500_000)
coordinate = (106.1, -6.2) 
rgb_red = (255, 0, 0)
````

perbedaan performa: tuple lebih cepat dari list untuk akses dan lebih hemat memory

````python
import sys

list_example = [1, 2, 3, 4, 5]
tuple_example = (1, 2, 3, 4, 5)

print(sys.getsizeof(list_example))  # Ukuran list dalam byte
print(sys.getsizeof(tuple_example))  # Ukuran tuple dalam byte
````

tuple bisa sebagai dictonary key, list tidak bisa

````python
location_data = {
    (106.1, -6.2): "Jakarta",
    (110.4, -7.8): "Yogyakarta"
}

print(location_data[(106.1, -6.2)])  # Mengakses nilai berdasarkan key tuple
````

## 3.3 Unpacking

````python
# Basic unpacking

employee = ("E001", "Andi Pratama", "Engineer", 7_500_000)
emp_id, nama, title, salary  = employee

print(emp_id)
print(title)
````

````python
# Swap variabel
x, y = 10, 20
x, y = y, x     # di bahasa lain butuh variabel sementara
print(x, y)     # 20 10
````

````python
# Extend unpacking dengan *
first, *midle, last = [1, 2, 3, 4, 5]

print(first)
print(midle)
print(last)
````

````python
# Mengabaikan nilai tertentu
_, name, _, salary  = ("E001", "Andi", "Engineer", 7_500_000)

print(name, salary)
````

Extended unpacking dengan * berguna untuk mengambil elemen pertama dan terakhir dari sequence tanpa tahu panjangnya

````python
def get_stat(data):
    return min(data), max(data), sum(data) / len(data)

minimum, maximum, average = get_stat([1, 2, 3, 4, 5])
print(f"Min: {minimum}, Max: {maximum}, Average: {average} \n")

# Tanpa unpacking hasilnya tuple
result = get_stat([1, 2, 3, 4, 5])
print(type(result))
print(result)
````

## 3.4 Named Tuple

Named tuple memberikan nama kesetiap posisi dalam tuple, membuat kode readable  

````python
from collections import namedtuple

# Definisi
Employee = namedtuple('employee', ['id', 'name', 'position', 'salary'])

# Membuat instance
emp = Employee(id="E001", name="Bagus", position="Data Science", salary=7_500_000)

# AKSES berdasarkan nama
print(emp.name)
````

````python
# Akses berdasarkan index
print(emp[1])

# Masih immutable
# emp.salary = 8_000_000  → AttributeError

# Membuat instance baru dengan satu field berbeda
emp2 = emp._replace(salary=8_000_000)
print(emp2.salary)    # 8000000

# Konversi ke dict
print(emp._asdict())
````
