# Floating Point Precission

````python
print(0.1 + 0.2)
````

hasil bukan 0.3 tetapi 0.30000000000000004 > karena angka desimal tertentu dapat dipresentasikan secara tepat dlaam format floating-point biner

````python
print(0.1 + 0.2 == 0.3)
````

````python
# Untuk kebutuhan tertentu

import math

math.isclose(0.1 + 0.2, 0.3) 

# == tidak selalu aman untuk membandingkan float, gunakan math.isclose() untuk membandingkan float
````

# F-string

* Formatted string literal: cara di python memasukan nilai variabel  atau hasil operasi langsung ke dalam sebuah string

````python
name = "John"
age = 30

print(f"My name is {name} and I am {age} years old.")
````

```
f"Halo, {name}!"
          ↑
      ambil nilai name
```

````python
price = 10000
quantity = 3

print(f"Total: {price * quantity}")
````

````python
name = "budi"

print(f"Nama: {name.upper()}")
````

````python
score = 0.85678

print(f"Score: {score:.2f}")
````


:.2f berarti:

* f → format sebagai floating-point
* .2 → tampilkan 2 angka di belakang koma

````python
mae = 19.322
rmse = 25.781
r2 = 0.89

print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.2%}")
````
