# 1. Iterable dan Iterator

## 1.1 Iterable vs Iterator

#### Iterable

* Definisi: objek yang dapat menyediakan elemen secara berurutan untuk diproses, misal: melalui for loop.
* Secara teknis: objek yang mengimplementasikan method `__iter__()`, yang mengembalikan sebuah iterator.
* Fungsi utama: menyediakan sumber data yang dapat ditelusuri satu per satu tanpa harus mengakses elemen secara manual.
* Contoh: list, tuple, dict, set, str, dan range.
* Karateristik: iterable tidak harus menyimpan posisi iterasi. objek iterable umumnya dapat menghasilkan iterator baru stiap kali `iter()` dipanggil.

````python
number = [10, 20, 30]

iterator = iter(number)

print(next(iterator))
````

- `number = [10, 20, 30]`
  - Membuat sebuah `list` berisi tiga bilangan bulat: `10`, `20`, dan `30`.
  - `number` merupakan objek iterable karena dapat menyediakan elemen untuk diiterasi.
- `iterator = iter(number)`
  - Fungsi `iter()` memanggil mekanisme `__iter__()` pada `number` untuk menghasilkan sebuah iterator.
  - Iterator digunakan untuk mengakses elemen list secara berurutan.
  - Pada tahap ini, iterator belum mengambil elemen apa pun; posisinya berada pada awal iterasi.
- `print(next(iterator))`
  - Fungsi `next()` meminta iterator menghasilkan elemen berikutnya.
  - Iterator menghasilkan nilai `10`, kemudian memperbarui state-nya agar iterasi berikutnya mengambil nilai `20`.
  - Fungsi `print()` menampilkan nilai yang dihasilkan ke layar.

Output:

```
10
```
