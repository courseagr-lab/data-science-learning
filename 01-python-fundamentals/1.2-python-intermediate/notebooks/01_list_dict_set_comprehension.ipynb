# 1. Konsep Comprehension

## 1.1 Konsep dan tujuan comprehension

### Comprehension

* Comprehension: Cara singkat untuk membuat collection baru dari sebuah iterable
* Comprehension menggabungkan tiga proses utama:
    - Iterasi > mengambil setiap elemen
    - Transformasi > Mengubah atau memproses elemen
    - Filtering > memilih elemen berdasarkan kondisi

* Jenis Comprehension:
    - List comprehension → menghasilkan list.
    - Set comprehension → menghasilkan set.
    - Dictionary comprehension → menghasilkan dict.

* Tujuan: membuat kode lebih ringkas dan mudah dibaca untuk operasi collection sederhana.

Bentuk umum sintak comprehension

`[expression for item in iterable if condition]`

Ketiga komponen tersebut memiliki peran yang berbeda:

*  `expression` — menentukan bagaimana setiap elemen ditransformasi sebelum dimasukkan ke collection baru.
* `for item in iterable` — menentukan sumber data yang diiterasi.
* `if condition` (opsional) — menentukan elemen mana saja yang lolos untuk diproses.

### Collection

* Collection adalah struktur data untuk menyimpan dan mengelompokkan beberapa nilai dalam satu objek.
* Collection memudahkan kita untuk menyimpan, mengakses, dan mengolah banyak data.
* Collection yang umum digunakan dalam Python:

  * `list` → menyimpan data secara berurutan dan dapat diubah.
  * `tuple` → menyimpan data secara berurutan dan tidak dapat diubah.
  * `set` → menyimpan data unik tanpa urutan tertentu.
  * `dict` → menyimpan data dalam pasangan `key-value`.

Contoh:

```python
numbers = [10, 20, 30, 40]
```

`numbers` adalah sebuah `list` yang berisi beberapa nilai.


### Contoh Comprehension

Contoh dasar: mengubah list bilangan menjadi list kuadratnya

````python
angka = [1, 2, 3, 4, 5]


# expression = num ** 2 (transformasi)
# for num in angka      (sumber iterasi)
kuadrat = [num ** 2 for num in angka]

print(f"Angka raw: {angka}")
print(f"Angak comprehension {kuadrat}")
````

### Kasus di Dunia Nyata

Dalam tahap data preprocessing, sering dijumpai data mentah berupa string yang perlu dikonversi ke tipe numerik sebelum dimuat ke dalam pandas.DataFrame. Misalnya, data harga produk hasil scraping web sering berbentuk string dengan simbol mata uang.

Data mentah hasil scraping — harga masih berupa string dengan simbol "Rp" dan titik ribuan

````python
harga_mentah = ["Rp15.000", "Rp23.500", "Rp8.000", "Rp42.000"]

harga_bersih = [
    float(harga.replace("Rp","").replace(".", ""))
    for harga in harga_mentah
]

print(harga_bersih)
````

### Interpretasi

* Comprehension bukan hanya cara untuk mengurangi jumlah baris kode.
* Comprehension menekankan pada **apa yang ingin dihasilkan**, bukan **bagaimana proses iterasinya dilakukan**.
* Cara berpikir ini disebut pendekatan **deklaratif**.
* Konsep ini relevan dalam Data Science karena mirip dengan penggunaan operasi **vectorized** di NumPy dan Pandas.
* Fokus utamanya adalah **hasil transformasi data**, bukan mekanisme loop.


## 1.2 Comprehension vs `for` loop

* Conprehension dan for loop sama sama digunakan untuk melakukan iterasi terhadap iterable.
* Comprehension biasanya digunakan ketika prosesnya sederhana seperti transformasi / filtering data.
* 'for` loop lebih cocok ketika proses membutuhkan beberapa langkah atau logika yang kompleks.
* Perbadaan utama terletak pada penyampaian intent:
    - Comprehension > fokus pada hasil yang ingin dibuat.
    - for loop > fokus pada langkah langkah proses.


Contoh for loop

````python

squers = [
]

numbers= [1, 2, 3, 4 ,5]

for x in numbers:
    squers.append(x**2)

print(squers)
````

Bentuk Comprehension

````python
squers_Com = [x ** 2 for x in numbers]

print(squers_Com)
````

* Keduanya menghasilkan hasil yang sama, tetapi comprehension lebih ringkas untuk operasi sederhana.
* Dari sisi performa, comprehension sering kali sedikit lebih cepat daripada for loop yang menggunakan append(), tetapi perbedaan ini biasanya bukan alasan utama memilih comprehension.

Pilih comprehension jika:
- Operasi sederhana.
- Hanya membutuhkan transformasi atau filtering.
- Hasil akhirnya berupa collection baru.


Pilih for loop jika
- Logika proses lebih kompleks.
- Membutuhkan beberapa statement.
- Membutuhkan kondisi atau proses yang sulit dibaca jika ditulis sebagai comprehension.

### Contoh

````python
data = [1, 2, 3, 4, 5, 6, 7, 8]

# ----- Pendekatan Iteratif -------------

hasil_loop=[]
for num in data:
    if num % 2 == 0: # filtering: hanya bilangan genap
        hasil_loop.append(num ** 2) # transformasi: kuadratkan


# --- Pendekatan deklaratif (comprehension) ---

hasil_comprehension = [num ** 2 for num in data if num % 2 ==0]

print(hasil_loop)         
print(hasil_comprehension)
````

Dari sisi performa, comprehension pada CPython umumnya sedikit lebih cepat dibandingkan for loop dengan list.append(), karena:

* Comprehension dikompilasi menjadi bytecode khusus (LIST_APPEND) yang dieksekusi dalam scope tersendiri.
* for loop dengan .append() melibatkan pemanggilan method secara eksplisit pada setiap iterasi, yang memiliki overhead lebih besar.

Namun demikian, for loop tetap lebih tepat digunakan ketika terdapat side effect (misalnya logging, penulisan ke file, atau multiple operations per iterasi) yang tidak menghasilkan collection baru secara langsung.

### Kasus di Dunia Nyata

Pada tahap pembersihan teks sebelum tokenisasi NLP, seluruh string pada suatu kolom sering perlu di-strip whitespace-nya secara seragam.

````python
kalimat_mentah = ["  saya suka data science  ", "python itu fleksibel   ", "  belajar ML"]

kalimat_bersih = [kalimat.strip() for kalimat in kalimat_mentah]

print(kalimat_bersih)
````

### Interpretasi

Kedua pendekatan bersifat functionally equivalent untuk kasus pembentukan collection baru. Pemilihan di antara keduanya sebaiknya didasarkan pada kompleksitas logika: gunakan comprehension untuk transformasi/filtering satu langkah yang jelas, dan gunakan for loop eksplisit ketika logikanya melibatkan banyak langkah atau side effect.

```text
Imperatif (for loop)                 Deklaratif (comprehension)
─────────────────────                ───────────────────────────
"Bagaimana caranya?"                 "Apa yang ingin dihasilkan?"
Eksplisit, step-by-step              Ringkas, satu ekspresi
Cocok untuk logika kompleks          Cocok untuk transformasi sederhana
Cocok untuk side-effect              Fokus pada collection baru


# 2. List Comprehension

## 2.1 Transformasi dan filtering data

### 2.1.1 Konsep Transformasi Data

* Transformasi adalah mengubah setiap elemen menjadi nilai baru
* Pada list comprehension, transformasi ditentukan oleh `ekspression`.
* Bentuk dasar:

    `[expression for item in iterable]`
* item mewakili setiap elemen dari `iterable`
* `expression` menentukan nilai mana yang akan dimasukan ke list baru


Contoh

````python
nilai = [80, 90, 76, 82]

nilai_baru = [nilai + 5 for nilai in nilai]

print(nilai_baru)
````

* Setiap nilai diproses satu per satu menggunakan nilai + 5.
* Jumlah elemen tetap sama karena tidak ada filtering.

### 2.1.2 Konsep Filtering Data

* Filtering : proses memilih elemen berdasarkan kondisi tertentu
* Pada list comprehension, filtering dilakukan menggunakan klausa if.
* Bentuk dasar

    `[item for item in iterable if condition]`

* `condition` menentukan apakah suatu elemen akan dimasukan kehasil.
* Jika kondisi `True` elemen dimasukan.
* Jika kondisi `False` elemen dilewati.

Contoh

````python
nilai_mtk = [40, 50, 65, 35]

nilai_lulus = [nilai for nilai in nilai_mtk if nilai > 60]

print(nilai_lulus)
````

### 2.1.3 Menggabungkan Filtering dan Transformasi

* Filtering dan Transformasi bisa dilakukkan dalam satu comprehension
* Polanya

    `[expression for item in iterable if condition]`

* `if condition` menentukan elemen mana yang akan diproses.
* `expression` menentukan nilai apa yang akan dimasukan kehasil.


Contoh

````python
nilai_ujian = [45, 67, 89, 32, 78, 90, 55, 100]

nilai_lulus = [nilai+5 for nilai in nilai_ujian if nilai > 60]
print(nilai_lulus)
````

Proses

```text

     ↓

  nilai >= 60

   ↙    ↘
 Ya     Tidak

ubah    dilewati
nilai
 ↓
hasil

### 2.1.4 Urutan Evaluasi

* Urutan evaluasi comprehension penting untuk dipahami.
* Untuk setiap item
    1. Python mengambil `item` dari `iterable`.
    2. Python mengevaluasi `condition`.
    3. Jika `condition` bernilai `True`, `expression` dievaluasi.
    4. Hasil `expression` dimasukan ke dalam collection baru.
    5. jika `condition` bernilai `True`, `exspression` tidak dijalankan.


Secara konseptual:

```text
for item
   ↓
if condition
   ↓
True?
 ├── Tidak → lewati item
 └── Ya
      ↓
 expression
      ↓
 masukkan ke hasil

### 2.1.5 Filtering Sebelum Transformasi

* Dalam pola

    `[expression for item in iterable if condition]`

* `if` berada setelah `for` tetapi secara logika filter diterapkan sebelum transformasi.
* Artinya, elemen yang tidak memenuhi kondisi tidak akan diproses oleh expression.
* Hal ini berbeda dengan melakukan transformasi terlebih dahulu lalu melakukan filtering terhadap hasil transformasi.

````python
hasil = [x**2 for x in numbers if x > 0]
print(hasil)
````

Secara konsep:

```text
ambil x
  ↓
x > 0 ?
  ↓
jika True
  ↓
x ** 2
  ↓
masukkan hasil

### 2.1.6 Interpretasi Fundamental

* `for` menentukan data yang akan diperiksa.
* `if` menentukan data yang dipilih.
* `expression` menentukan bentuk data hasil.
* Karena filtering dilakukan sebelum `expression`, elemen yang ditolak tidak mengalami transformasi.
* Pola ini menjadi dasar untuk memahami comprehension yang lebih kompleks.

### Kasus di Dunia Nyata

Pada tahap exploratory data analysis (EDA) awal, seringkali diperlukan filtering sederhana terhadap outlier ekstrem sebelum data divisualisasikan, atau mengekstraksi kolom tertentu dari sekumpulan record berbentuk dictionary.

````python
# Data transaksi sederhana dalam bentuk list of dict
transaksi = [
    {"id": 1, "jumlah": 150000},
    {"id": 2, "jumlah": 2500000},   # outlier
    {"id": 3, "jumlah": 89000},
    {"id": 4, "jumlah": 3000000},   # outlier
    {"id": 5, "jumlah": 120000},
]

# Filtering transaksi wajar (di bawah 1 juta) sekaligus ekstraksi nilai "jumlah"
jumlah_wajar = [t["jumlah"] for t in transaksi if t["jumlah"] < 1_000_000]

print(jumlah_wajar)
````

## 2.2 Conditional Expression pada Comprehension

### 2.2.1 Konsep Conditional Expression

* Conditional expression digunakan untuk menentukan nilai hasil berdasarkan kondisi.
* Berbeda dengan filtering, conditional ekspression.
* Setiap elemen tetap menghasilkan satu nilai.
* Bentuk dasar:

    ```python
    [nilai_jika_true if kondisi else nilai_jika_false for item in interable]
    ```

* Conditional expression ditempatkan pada bagian expression, yaitu sebelum for



### 2.2.2 Conditional Expression vs Filtering

* Filtering menggunakan `if` untuk menentukan elemen yang dipertahankan.
* Conditional expression menggunakan `if-else` untuk menentukan nilai yang dihasilkan.

| Filtering                            | Conditional expression           |
| ------------------------------------ | -------------------------------- |
| Menentukan elemen yang dipertahankan | Menentukan nilai yang dihasilkan |
| Beberapa elemen dapat dibuang        | Semua elemen tetap diproses      |
| `if` berada setelah `for`            | `if-else` berada sebelum `for`   |


Contoh filtering:

```python
[x for x in numbers if x > 0]
```

Contoh conditional exspression:

```python
["positif" if x > 0 else "non_positif" for x in numbers]

### 2.2.3 Alur Evaluasi

untuk setiap `item`

```text
        item
          ↓
      evaluasi kondisi
          ↓
     ┌────┴────┐
   True       False
     ↓           ↓
nilai_true   nilai_false
     └────┬────┘
          ↓
    masuk ke hasil
```

* Tidak ada elemen yang dilewati
* Setiap `item` menghasilkan tepat satu nilai.

### 2.2.4 Contoh Klasifikasi Sederhana

````python
suhu_harian = [18, 25, 30, 15, 33, 22]

kategori_suhu = [
    "Panas" if suhu > 25 else "Dingin" for suhu in suhu_harian
]


import pandas as pd

data_suhu = pd.DataFrame({
    "Suhu":suhu_harian,
    "Kategori":kategori_suhu
})

print(data_suhu)
````

### 2.2.5 Conditional Expression Bertingkat

* Conditional expression dapat lebih memiliki satu kondisi
* Contoh:

````python
kategori_suhu_1 = [
    "panas" if suhu > 30 
    else "dingin" if suhu <= 18
    else "normal"
    for suhu in suhu_harian
]

print(kategori_suhu_1)
````

Alurnya:

```text
suhu
  ↓
suhu >= 30 ?
 ├─ True  → "panas"
 └─ False
       ↓
   suhu <= 18 ?
    ├─ True  → "dingin"
    └─ False → "normal"

### Kasus di Dunia Nyata

Salah satu praktik umum sebelum encoding kategorikal adalah binning — mengubah nilai numerik kontinu menjadi label kategori berdasarkan ambang batas (threshold) tertentu.

````python
skor_risiko = [0.12, 0.65, 0.45, 0.90, 0.30]

# Binning sederhana: risiko tinggi jika skor > 0.5, selain itu risiko rendah
label_risiko = ["tinggi" if skor > 0.5 else "rendah" for skor in skor_risiko]

print(label_risiko)  # ['rendah', 'tinggi', 'rendah', 'tinggi', 'rendah']
````

### Interpretasi

Kesalahan yang umum dilakukan pemula adalah tertukar antara filtering if (di akhir ekspresi, memfilter elemen) dan conditional expression if-else (di awal ekspresi, mengubah nilai tanpa memfilter). Prinsip pembeda utamanya:

* Filtering if → jumlah elemen hasil bisa lebih sedikit dari input.
* Conditional if-else → jumlah elemen hasil selalu sama dengan input, hanya nilainya yang berbeda.
* Filtering menjawab: "Elemen mana yang dipertahankan?"
* Conditional expression menjawab: "Nilai apa yang diberikan kepada setiap elemen?"
*  Ini adalah perbedaan fundamental yang penting saat membaca comprehension.

Kedua pola ini bahkan dapat digabungkan dalam satu comprehension apabila diperlukan filtering sekaligus percabangan nilai.

# 3. Dict & Set Comprehension

## 3.1 Dictionary Comprehension

### 3.1.1 Konsep Dasar

* Dictionary comprehension > cara singkat membuat dict dari sebuah iterrable.
* Setiap elemen digunakan untuk membentuk pasangan `key-value`
* Bentuk umum:

    ```python
    {key_expression: value_expression for item in iterable}
    ```

* `key_expression` menentukan key.
* `value_expression` menentukan value.
* Hasil akhir berupa dictionary. 

### 3.1.2 Dictionary Comprehension dengan `zip()`

* `zip()` digunakan untuk memasangkan elemen berdasarkan posisinya.
* Contoh:

````python
produk = ["laptop", "mouse", "keyboard", "monitor"]
harga = [12000000, 150000, 450000, 2500000]

katalog_harga = {
    nama:hrg
    for nama, hrg in zip(produk, harga)
}

print(katalog_harga)
````

* `zip(produk, harga) menghasilkan pasanganL

```text
("laptop", 12000000)
("mouse", 150000)
("keyboard", 450000)
("monitor", 2500000)

* Dictionary comprehension kemudian membentuk:

```python
("laptop", 12000000)
("mouse", 150000)
("keyboard", 450000)
("monitor", 2500000)

### 3.1.3 Alur Konseptual

```text
produk + harga
      ↓
    zip()
      ↓
 pasangan (nama, harga)
      ↓
    iterasi
      ↓
 nama → key
 harga → value
      ↓
 dictionary baru
```

* Dictionary berguna untuk membuat mapping atau lookup table dari data yang sudah memiliki hubungan key-value.
* Filtering juga bisa ditambahkan `if`

     ```python
     {key_expression: value_expression for item in iterable if condition}
     ```

* Prinsip utama: iterasi > tentukan key > tentukan value > bentuk dictionary.

### Kasus di Dunia Nyata

Dict comprehension sering dipakai untuk membangun term-frequency sederhana atau mapping index-ke-label, misalnya saat melakukan manual encoding pada label kategorikal sebelum digunakan dalam model machine learning.

````python
kelas_unik = ["setosa", "versicolor", "virginica"]

# Membangun mapping label > index numerik (label encoding manual)
label_ke_index = {label: idx for idx, label in enumerate(kelas_unik)}

print(label_ke_index)
````

### Interpretasi

Dict comprehension mengekspresikan transformasi struktur data secara simetris antara list comprehension dan dictionary — konsepnya identik, hanya berbeda pada bentuk output (pasangan key-value, bukan elemen tunggal). Pola ini menjadi dasar dari banyak fungsi encoding/decoding yang akan sering ditemukan pada pipeline preprocessing.

## 3.2 Set Comprehension

### 3.2.1 Konsep Dasar

* Set comprehension: cara ringkas untuk membuat `set` baru dari sebuah `iterable`.
* Set menyimpan elemen unik, sehingga duplikat dihilangkan.
* Bentuk umum:

    ```python
    {expression for item in iterable if condition}
    ```

* `expression` menentukan nilai yang dimasukan ke dalam list.
* 'if condition` bersifat opsional dan digunakan untuk filtering.
* Bentuk umum:

    ```python
    {expression for item in iterable if condition}
    ```

* `expression` menentukan nilaiyang dimasukan ke dalam set.
* `if condition` bersifat opsional dan untuk filtering.
* Hasil akhir berupa set.


### 3.2.2 Transformasi dan Penghapusan Duplikat

````python
kata = ["data", "science", "data", "python", "science", "model"]

kata_unik = {k.upper() for k in kata}
print(kata_unik)
````

Alur:

```text
kata
  ↓
iterasi setiap kata
  ↓
k.upper()
  ↓
ubah menjadi huruf kapital
  ↓
masukkan ke set
  ↓
duplikat otomatis dihilangkan

Hasil:

```text
{{'DATA', 'SCIENCE', 'PYTHON', 'MODEL'}
```
* "data" muncul dua kali, tetapi "DATA" hanya disimpan satu kali.
* "science" juga hanya menghasilkan satu "SCIENCE".
* Set tidak menjamin urutan elemen saat ditampilkan.}

### 3.2.3 Set Comprehension dengan Filtering

filtering dapat ditambahkan setelah `for`

````python
kata_panjang = {
    k.upper()
    for k in kata
    if len(k) > 4
}

print(kata_panjang)
````

* `for k in kata` → melakukan iterasi.
* `if len(k) > 4` → memilih kata dengan panjang lebih dari 4.
* `k.upper()` → mengubah kata menjadi huruf kapital.
* {...} → menghasilkan set sehingga duplikat tetap dihilangkan.

### 3.2.4 Interpretasi

* Set comprehension digunakan ketika membutuhkan transformasi data sekaligus memastikan hasilnya unik.
* Berbeda dengan list comprehension, hasil set tidak mempertahankan duplikat.
* Pola fundamentalnya:
    
    ```text
    iterasi → filtering (opsional) → transformasi → set unik

### Kasus di Dunia Nyata

Set comprehension sangat berguna untuk mengekstraksi nilai unik dari data mentah, misalnya mengambil domain email yang berbeda dari sekumpulan alamat email pengguna untuk keperluan analisis kategori sederhana.

````python
email_pengguna = [
    "budi@gmail.com", "siti@yahoo.com", "andi@gmail.com",
    "rina@outlook.com", "dedi@yahoo.com"
]

# Ekstrak domain unik menggunakan split() lalu set comprehension
domain_unik = {email.split("@")[1] for email in email_pengguna}

print(domain_unik)
# Output: {'gmail.com', 'yahoo.com', 'outlook.com'}
````

# 4. Nested Comprehension


## 4.1 Comprehension pada data dua dimensi

### 4.1.1 Konsep Nested Comprehension

* Nested comprehension: comprhension yang menggunakan lebih dari satu klause `for`.
* Digunakan untuk memproses data bertingkat:
    - matrix
    - `list of list`
    - data dengan struktur bersarang.
* Urutan `for` mengikuti `for` loop bersarang biasa.
* `for` pertama  outer loop.
* `for` berikutnya = inner loop


Bentuk umum:

```python
[expression for item_outer in iterable_outer for item_inner in item_outer]


### 4.1.2 Flatten Data Dua Dimensi

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

flat = [elemen for baris in matrix for elemen in baris]

print(flat)
````

* `for baris in matrix` > mengambil setiap baris.
* `for elemen in baris` > mengambil setiap elemen dari baris tersebut.
* `elemen` > nilai yang dimasukan ke dalam list.

### 4.1.3 Urutan Evaluasi

Contoh urutan:

```text
matrix
   ↓
for baris in matrix
   ↓
ambil satu baris
   ↓
for elemen in baris
   ↓
ambil setiap elemen
   ↓
masukkan elemen ke hasil
   ↓
lanjut ke baris berikutnya

```text
baris [1, 2, 3]
    ↓
1 → 2 → 3

baris [4, 5, 6]
    ↓
4 → 5 → 6

baris [7, 8, 9]
    ↓
7 → 8 → 9

### 4.1.4 Kesetaraan dengan Nested for Loop

Comprehension

````python
flat = [elemen for baris in matrix for elemen in baris]

print(flat)
````

Setara dengan

````python
flat_1 = []

for baris in matrix:
    for elemen in baris:
        flat_1.append(elemen)

print(flat_1)
````

* `for` pertama pada comprehension menjadi outer loop.
* `for` kedua menjadi inner loop.
* `expression` (`elemen`) menjadi nilai yang dimasukan ke hasil.
